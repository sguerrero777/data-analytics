### `Table View`

**Notice the icons next to some of the fields in the table. What do you think each 
means?**

Different icons indicate the data type or behavior of each field:
- **Σ (Sigma)** = Numeric field that can be aggregated (sum, average, etc.)
- **Calendar** = Date/time field
- **ABC** = Text field
- **Calculator/chart icon** = A calculated measure (DAX formula)

**Notice the icon that looks like a crossed off eye next to only some of the fields. What 
do you think that means?**

The crossed-off eye icon means the field is **hidden from Report View**. It still
exists in the data model and can be used in calculations or relationships, but it
won't appear as a draggable field when building visuals. This is typically applied
to technical columns like ID/foreign key fields that aren't meaningful to end users.

### `Models View`

**What table is at the center?**

The table at the center is the Fact table. This is the table that holds all the numeric measures like profit margins, revenue, and expenses, along with the foreign keys that point outward to all the surrounding tables.

**For tables branching out from the center, do they connect to other tables?**

Yes, some tables do connect to other tables. For instance, the Customer table branches out from the Fact table, then the State table branches out from the Customer table through the shared Postal Code foreign key.

**What kind of shape does this layout seem to make?** 

This layout seems similar to an ERD I've seen built in MySQL Workbench. Because some of the outer tables also connect to other outer tables (like Customer → State), it's more of a snowflake shape.

**What kind of information shows up in the center table, compared to the outer 
tables?**

Compared to the other tables, the center table includes the numeric measures such as profit margins, revenue, and expenses,| along with foreign keys that reference the outer 
tables. The outer tables hold more descriptive information, like names, categories, and geographic details.
