<h2 style="color:steelblue">Section 2.A: The Data Modeling Process</h2>

### 2.A.1 Modeling Basics

A **data model** describes the shape and relationships within your data — how each part relates to the others.

| Type | What It Covers |
|---|---|
| **Simple Modeling** | Just table shape and data types — what relational DBs offer |
| **Advanced Modeling** | Advanced calculations, explicit relationship rules, attribute hierarchies, data formatting preferences |

> Modeling = shaping data in a way that works best for **YOUR** needs — different needs mean different models.

### 2.A.2 & 2.A.3 Modeling Stages

There are **6 stages** to the modeling pipeline:

| # | Stage | What You Do |
|---|---|---|
| 01 | **Discover** | Learn the existing shape of your data, understand what values are present, identify what's useful — use a Data Dictionary and EDA |
| 02 | **Bring In (Ingest)** | Verify you have security rights to access the data, determine the time window for the data pull |
| 03 | **Profile** | Examine the source data, verify it matches expectations, note any anomalies |
| 04 | **Transform** | Combine/split tables, rename columns and tables, change data types |
| 05 | **Clean** | Remove rows/columns with irreparable issues, fix errors, fill gaps, standardize formats |
| 06 | **Augment** | Add calculations (simple: single row or basic lookup / advanced: aggregate across many rows) and add attribute intelligence (define hierarchies, specify date attributes) |

**Full pipeline:** Discover → Ingest → Profile → Transform → Clean → Augment

> **Transform Note:** Power BI supports human-oriented data types like **Percentage** 
> and **Currency** — beyond the Decimal/Float of traditional databases.

### 2.A.4 Modeling Is Transforming & Describing

Modeling = taking a **Source Model** and turning it into a **Destination Model** through a series of steps.

| | Source Model | Destination Model |
|---|---|---|
| **What it is** | Raw data as it exists in the source system | Optimized for analysis in Power BI |
| **Examples** | Flat files, OLTP DBs, APIs, JSON, XML | Star/Snowflake Schema, Semantic Model, DAX-ready |

**Transformation steps in between:**
- Profile source data
- Reshape tables
- Rename columns
- Clean bad data
- Add calculations
- Define relationships

>  Understanding your **source** data is essential.  
> Understanding your ideal **destination** model is equally essential.

### 2.A.5 Source Models

Understanding where your data comes from helps define your starting point. Source data can come from:

| Source Type | What It Is |
|---|---|
| **Flat Files** | CSV, Excel, TSV — simple row/column structure, no built-in relationships |
| **OLTP Databases** | Online Transactional Processing — highly normalized, optimized for fast writes |
| **OLAP Databases** | Online Analytical Processing — star/snowflake schema, ideal for reporting |
| **APIs** | REST or SOAP endpoints — structured JSON/XML responses from external services |
| **JSON Files** | JavaScript Object Notation — nested, semi-structured, common in web services |
| **XML Files** | Extensible Markup Language — hierarchical, tag-based, common in enterprise systems |

### 2.A.6 Relational Databases: OLTP vs OLAP

Both are relational databases but built for completely different purposes:

| | **OLTP** | **OLAP** |
|---|---|---|
| **Full Name** | Online Transactional Processing | Online Analytical Processing |
| **Purpose** | Handle many fast, simple transactions | Structured for reporting and analysis |
| **Normalization** | Highly normalized | Less normalized |
| **Schema** | No standard schema | Star / Snowflake schema |
| **Also Called** | Transactional DB | Data Warehouse |
| **Example Use** | A cash register recording every sale | A dashboard showing monthly sales trends |

> In OLTP, relationships between tables are implied through **foreign keys** and 
> re-specified with every SQL JOIN. In Power BI, relationships are **explicit** and 
> defined once in the model.

### 2.A.7 An Ideal Power BI Model

The ideal Power BI structure is a **Star or Snowflake Schema** — clear fact + dimension 
table structure, no circular references.

>  OLAP/Data Warehouse sources match this perfectly — but they are **rarely available** in practice.

In practice, you usually get one of two messy extremes:

| | **Broken Into Many Tables** | **Flattened Into One Table** |
|---|---|---|
| **Typical source** | OLTP databases | Flat file exports |
| **Problems** | Large number of highly normalized tables, relationships may form circular loops, needs reshaping before use in Power BI | All data in one table, relationship data is lost, must infer groupings manually, harder to build flexible reports |

>  **Power BI's job:** Take these imperfect sources and shape them into a clean 
> Star/Snowflake semantic model.

<h2 style="color:steelblue">Section 2.B: Power BI Modeling</h2>

### 2.B.1 Power BI Models

In Power BI, a data model is called a **Semantic Model** — powered by the 
**xVelocity Engine** (the same engine used in Excel Power Pivot and SQL Server 
Analysis Services).

Modeling in Power BI Desktop happens across four tasks:

- **Import Data** — using Power Query M to connect to and load data sources
- **Configure Relationships** — using Model View to define how tables relate
- **Data Types & Preferences** — setting column types, formats, categories, and sort orders
- **Write DAX Expressions** — creating calculated columns and measures for analysis

> The xVelocity engine is the same in-memory analytical engine underlying 
> Excel Power Pivot — mastering one means familiarity with the others.

### 2.B.2 Table Modeling

Power BI models are built around two types of tables — the same concept as 
OLAP/Data Warehouse design:

- **Data Tables** (also called Fact Tables) — sit at the center of the model
- **Lookup Tables** (also called Dimension Tables) — surround the data tables

> These two table types working together form the foundation of a 
> Star or Snowflake schema in Power BI.

### 2.B.3 Star & Snowflake Schema Rules

> It is up to **me** to enforce these rules — most difficult data analysis 
> problems are rooted in a flawed model.

**Rule 1 — Data ↔ Lookup Only**  
Data (fact) tables should relate ONLY to lookup (dimension) tables. Never relate 
two data tables directly to each other.

**Rule 2 — Snowflake Exception**  
Lookup tables may relate to another lookup table only when following Snowflake 
Schema rules — a lookup normalised into sub-lookups.

**Rule 3 — Many-to-One Relationships**  
Relationships between data and lookup tables should be many (data records) to 
one (lookup record). Example: many sales rows → one product.

> Think of it as one-way traffic — data flows from lookup tables INTO data 
> tables through relationships.

### 2.B.4 Model Relationships

Relationships in Power BI are defined **explicitly** — unlike SQL where you 
re-specify joins in every query.

Every relationship has three properties:

**Cardinality** — the numerical relationship between rows on each side
- *Many-to-One* (most common): many rows in the data table match one row in the lookup
- *One-to-One*: each row on both sides matches exactly one row on the other
- *Many-to-Many*: avoid where possible — can cause ambiguous results

**Filter Direction** — which way filters flow across the relationship
- *Single*: filters flow from the lookup table into the data table (recommended)
- *Both*: filters flow both ways — use with caution, can cause unexpected results

**Granularity** — the level of detail at which data is stored on each side
- Example: an Orders table stores one row per order; an Order Details table 
  stores one row per line item — these are at different granularities

> Always consider granularity when connecting tables — mismatched 
> granularity causes incorrect aggregations in your visuals.

### 2.B.5 & 2.B.7 Data Tables vs. Lookup Tables

| | **Data Table (Fact)** | **Lookup Table (Dimension)** |
|---|---|---|
| **Also known as** | Fact table | Dimension table |
| **Contains** | Measures — numeric values to summarise (e.g. quantity sold, revenue) | Descriptive attributes — ways to describe, group, or filter facts (e.g. product name, region) |
| **Each record** | Describes an event or transaction — something that happened | Describes a noun — a person, place, product, or concept |
| **Schema position** | Center table of a star schema | Outer tables of a star schema |
| **Relationships** | Connects outward only to lookup tables, never to other data tables | Connects inward to data tables; may connect to other lookups in a snowflake schema |

> Together, data and lookup tables form the semantic model — the foundation 
> for all DAX calculations and report visuals.

### 2.B.6 & 2.B.8 Table Naming Conventions

Naming choices appear directly in report visuals and Natural Language Q&A — 
get them right from the start.

**Data Table Naming (2.B.6)**
- Use **plural** names representing the event: `Sales`, `Purchases`, `Shifts`, `Logins`
- Do NOT prefix with `Fact_`: ~~FactSales~~ → `Sales`, ~~FactPurchases~~ → `Purchases`

**Lookup Table Naming (2.B.8)**
- Use **singular** names: `Product`, `Employee`, `Currency`, `Customer`
- Do NOT prefix with `Dim_` or `Dimension_`: ~~DimProduct~~ → `Product`, ~~DimCustomer~~ → `Customer`

> ⚠️ Both `Fact_` and `Dim_` prefixes look out of place in Power BI report 
> visuals and break Natural Language Q&A queries.

### 2.B.9 Columns

Columns are the building blocks of every table. Power BI provides rich 
configuration options for each column:

- **Data Type** — every column has a type (number, text, date, boolean, binary). 
  Choosing the right type enables correct calculations and formatting.
- **Formatting** — currency symbols, decimal places, thousands separators, date 
  patterns — controls how values appear in visuals.
- **Sort Order** — a column can be sorted by a *different* column. 
  Example: sort Month Name by Month Number so January comes before July, not alphabetically.
- **Default Summarization** — controls the default aggregation when a numeric 
  column is dragged into a visual: Sum, Average, Count, Min, Max, or None.
- **Groups & Bins** — categorical data can be grouped; numeric data can be 
  placed into bins (ranges) for histogram-style analysis.

### 2.B.10 & 2.B.11 Column Naming Guidelines

Column names show up directly in report visuals — clarity and consistency matter.

- **Avoid Ambiguity — Add Descriptors**: instead of just `City`, use `Supplier City`, `Order Ship-To City`, `Customer City`
- **Remove Obvious Words**: `Customer Name` → `Customer`, `Shipping Company` → `Shipper`
- **Qualify Numeric Values**: add Amount or Percent — `Discount Amount`, `Discount Percent`, `Tax Amount`
- **Use Spaces & Full Words**: do not truncate or abbreviate — `Order Quantity`, not `OrdQty`
- **Use Correct Business Terminology**: `Net Sales Amount` is accepted; `Total Sales Value` is ambiguous — total of what?
- **Consistency & Readability**: use the same terms across all reports, document them, and ask peers for feedback

>  Quick Measures, templates, and examples found online are excellent 
> sources of proven naming conventions.

### 2.B.12 – 2.B.14 Column Datatypes & Formatting


| Category | Data Types | Formatting Options |
|---|---|---|
| **Numeric** | Decimal Number, Fixed Decimal Number, Whole Number | Currency, General number, Percentage, Thousands separator, Scientific notation, Decimal places |
| **Date / Time** | Date/Time, Date, Time, Date/Time/Timezone, Duration | Custom patterns (e.g. `YYYY-MM-DD`), Short/Long date, 12- or 24-hour time, Locale-specific |
| **Other** | Text, True/False, Binary | Text: no numeric formatting · True/False: boolean flags · Binary: file or image storage |

### 2.B.15 Column Data Categories

Data categories tell Power BI what kind of real-world information a column 
represents — enabling smarter visuals like maps.

- **Geography** — Country, State/Province, City, Postal Code, Latitude, Longitude
- **Hyperlink** — columns containing URLs that render as clickable links in visuals
- **Barcode** — columns used for barcode scanning in the Power BI mobile app
- **Image URL** — columns containing links to images to render them inline

> Without a data category, Power BI treats `Country` as plain text and won't 
> plot it on a map. Assigning the Geography category unlocks map visuals automatically.

### 2.B.16 Column Sort Orders

Columns can be configured to sort by a *different* column — essential for 
logical ordering in visuals.

**How it works:** In the Data/Table view, select the display column → click 
'Sort by Column' in the ribbon → choose the sort column.

| Display Column | Sort By Column | Result |
|---|---|---|
| `Month Name` | `Month Number` | Jan, Feb, Mar... instead of Apr, Aug, Dec... |
| `Full Name (First Last)` | `Last Name` | Adams, Brown... instead of Alice, Bob... |
| `Age Bracket` | `Age Bracket Order` | Infant, Child, Teen, Adult... instead of alphabetical |

>  Month Name sorted by Month Number is the most common use case — without 
> it, months display in alphabetical order in every visual.

### 2.B.17 Default Summarization

Controls the implicit measure created when a numeric column is dragged into a visual.
Default is **Sum** — but this should be changed when Sum doesn't make sense.

Example: a Popularity column (0–5 rating) should default to **Average**, not Sum.

| Type | What It Does |
|---|---|
| **Sum** | Total of all values — default for most numeric columns |
| **Average** | Mean value — better for ratings, scores, temperatures |
| **Count** | Number of rows — counts blank and non-blank |
| **Count (Distinct)** | Unique non-blank values only |
| **Min** | Lowest value in the group |
| **Max** | Highest value in the group |
| **None** | Disables automatic aggregation — useful for ID columns |

> 💡 You can always override summarization within a specific visual — the 
> default just controls the starting point when you drag a field in.

### 2.B.18 Navigational Hierarchies

A navigational hierarchy groups columns into a drill-down path that users 
can explore directly in report visuals. Defined in Model View by dragging 
one column onto another.

**Natural hierarchies** — follow obvious parent-child relationships:
- Date: `Year` → `Quarter` → `Month` → `Day`
- Geography: `Country` → `Region` → `County` → `City`

**Non-natural hierarchies** — no obvious order; typically go from fewest 
to most options, depending on your visualization objectives:
- `Gender` → `Marital Status` → `Age Bracket`
- `Category` → `Size` → `Color`

>  Power BI auto-detects date hierarchies when a Date data type is assigned. 
> Custom hierarchies must be defined manually in Model View.

<h2 style="color:steelblue">Section 2.C: Data Transformation</h2>

### 2.C.1 Transformation Pipelines

When building a new model, transformation happens in stages using three 
primary tools in Power BI Desktop:

**Power Query (M Language)**
- Primary transformation tool in Power BI
- Connects to sources and shapes data *before* it enters the model
- Step-based pipeline — every action is recorded and repeatable
- Handles renaming, type changes, filtering, merging, pivoting
- Runs automatically on each data refresh
- *Use for: structural transformations — shaping how data arrives*

**GUI Configuration**
- Settings applied through the Power BI Desktop interface
- Column data types, formatting, and categories
- Table and column naming
- Sort-by-column assignments
- Default summarization and navigational hierarchies
- *Use for: model-level metadata — how columns behave in visuals*

**DAX Expressions**
- Data Analysis eXpressions — Power BI's formula language
- Creates calculated columns (row-by-row values stored in the model)
- Creates measures (aggregations evaluated at query time)
- Enables complex business logic not possible in Power Query
- References other columns and tables within the semantic model
- *Use for: analytical enrichment — business logic and KPIs*

### 2.C.2 Transformation Order

The order of transformations varies by situation, but following this sequence 
minimises errors and rework:

1. **Bring In Raw Data** — load source data into a staging table, don't modify source files
2. **Remove Unneeded Columns** — drop columns not required for analysis, reduces model size and load time
3. **Remove / Repair Bad Data** — fix or remove rows that violate expected data types or ranges
4. **Remove Out-of-Scope Data** — filter rows outside the time range, geography, or business scope
5. **Define Column Types** — assign correct data types: numbers, dates, text, booleans
6. **Define Column Names** — apply naming conventions — clear, business-friendly, no abbreviations
7. **Fill Unknowns / Nulls** — replace null or blank values with defaults, lookups, or calculated substitutes
8. **Add Row-Level Calculations** — row totals, formatted variations, parent-child derivations, child aggregates

> Think of this as a funnel — get the shape right first, then worry about 
> names and values, then enrich.

### 2.C.3 Cleaning Column Data

Data cleaning corrects values that are wrong, inconsistent, or impossible — 
before they corrupt your analysis.

**Types of data issues to look for:**

- **Address / Spelling Correction** — fix city name variations: `'New Yrok'` → `'New York'`, `'Los Ang.'` → `'Los Angeles'`
- **Inconsistent Format / Mixed Types** — same field with different units or formats:
  `4`, `16`, `"12 year"`, `2011` all as answers to "how many years have you been teaching?"
- **Implausible Values** — values outside realistic bounds: `Age = 999`, impossible dates, negative quantities
- **Obvious Typos** — text errors that are clearly wrong: `'Malf'` instead of `'Male'`, `'Femal'` for `'Female'`

> Always be careful about making assumptions — verify before changing data, 
> and flag for review when uncertain.

### 2.C.4 Filling Unknowns with Lookups

When your data is missing context, enrich it by joining to external or 
reference datasets. This is done in Power Query using **Merge Queries** — 
essentially a JOIN that brings in columns from a reference table.

Examples of what you can pull in:

- **Demographic Information** — average income for a city/ZIP, school district by ZIP, population density by region
- **Economic Data** — interest rates over a period, unemployment rates, CPI/inflation rates by month
- **Weather Data** — annual/monthly rainfall per city, average temperature by location, historical weather events
- **Geological / Geographic Detail** — feet above sea level, population of a region, rural vs. urban classification by ZIP

### 2.C.5 Augmenting Column Data

Augmentation adds new derived columns that make data richer and more useful 
for reporting — going beyond what the source provides.

**Formatted Variations** — create display-friendly versions of existing columns
- Example: `FirstName & " " & LastName` → `"Jane Smith"` (a Display Name column)

**Pulling Parent Data into Children** — embed a parent attribute directly in 
the child table (denormalize)
- Example: `Products[Category Name]` ← `Category[Name]` so the Products table 
  stores Category Name instead of just a Category ID

**Pulling Child Aggregates into Parents** — summarize child records and store 
the result in the parent row
- Example: `Orders[Order Total] = SUM(LineItems[Amount])`
- Example: Employee table gets a `[# Sales Regions]` column

**Discretized Fields** — convert a continuous numeric value into a labelled 
category, making data easier to group, filter, and visualize
- Example: `Age` → `Age Bracket`:
  - Free Child (3 and younger), Child (4–12), Teen (13–18)
  - Young Adult (19–24), Adult (25–59), Senior (60+)
- Discretized fields enable drill-down hierarchies, slicer filters, and 
  grouping in visuals — transforming analysis from 'what' to 'who and when'

> Augmentation turns a basic data load into an analysis-ready model — 
> each new column should serve a clear reporting or calculation purpose.

## 🗝️ Module 2 | Key Takeaways

Here's everything that matters from this module in plain English:

**Modeling is just shaping your data to fit your needs.**  
There's no single "correct" model — it depends on what questions you're trying to answer.

**You'll almost never start with perfect data.**  
Real sources are either too broken up (OLTP) or too flat (CSV exports). Power BI's job 
is to reshape them into a clean Star or Snowflake schema.

**Star vs Snowflake — what's the difference?**  
- ⭐ **Star Schema** — data tables connect directly to lookup tables. Simple, flat, fast.  
- ❄️ **Snowflake Schema** — some lookup tables connect to *other* lookup tables 
  (sub-lookups). More normalized, slightly more complex.  
- Example: `Sales` → `Product` → `Category` is a snowflake because Product connects 
  to another lookup (Category) instead of storing Category Name directly.

**Two table types, one simple rule:**  
- 🟣 **Data tables** (plural names) — store *what happened*: Sales, Orders, Logins  
- 🔵 **Lookup tables** (singular names) — store *who/what was involved*: Product, Customer, Employee  
- Data tables connect to lookup tables. Never data-to-data, never lookup-to-lookup 
  (unless following snowflake rules).

**Relationships need three things: cardinality, filter direction, and granularity.**  
Get these wrong and your visuals will silently show incorrect numbers.

**Columns are more than just data — configure them properly:**  
Right data type → right formatting → right name → right sort → right summarization.  
Each setting affects how Power BI displays and calculates in your reports.

**Transformation has a tool for everything:**  
- 🔧 **Power Query** — shape and clean *before* data enters the model  
- 🖱️ **GUI Config** — set metadata like names, types, and sort orders  
- 📐 **DAX** — add business logic and calculations *after* data is in the model

**Always transform in order — don't skip steps.**  
Bring in → Remove junk → Clean bad data → Type it → Name it → Fill blanks → Augment.  
Skipping steps early creates problems that are painful to fix later.

**Cleaning isn't just typos.**  
Mixed types, implausible values, and inconsistent formats all silently break aggregations.  
When in doubt — flag it, don't assume.

**Augmentation = making your data smarter.**  
Add display names, pull in parent/child data, and discretize continuous values into 
categories so your reports can group, filter, and drill down meaningfully.
