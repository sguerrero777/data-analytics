<h2 style="color:steelblue">Section 3.A: Power Query</h2>

### 3.A.1 Power Query

Power Query is the primary data ingestion and transformation tool in Power BI — 
designed to be approachable to non-coders while supporting over 700 transformations 
across a diverse set of data sources.

**Power Query Editor (GUI)**
- Graphical interface for building queries without writing code
- Wizards and dialogs connect to 100+ data sources
- Ribbon tools perform common transformations with a click
- All actions recorded as named, editable steps

**M Language (Under the Hood)**
- Power Query Formula Language — the code behind every step
- Auto-generated as you interact with the GUI
- Advanced Editor allows direct M code authoring
- Referenced as 'M' or 'M Language' throughout the course

**700+ Transformations available including:**
- Filter, sort, group, and aggregate rows
- Split, merge, rename, and remove columns
- Pivot, unpivot, and transpose tables
- Parse text, dates, JSON, XML, and more

**Advanced Editor**
- Write or edit M expressions directly
- Useful for complex logic not exposed in the GUI
- Typically start with GUI-generated code, then modify
- Enables reuse and parameterization of queries

### 3.A.2 Power Query Steps

Every action in Power Query is recorded as a named step. Together, steps form 
a transformation pipeline that is repeatable and editable.

**A typical step pipeline might look like:**

`Source` → `Navigation` → `Remove Columns` → `Filter Rows` → `Changed Type` → `Renamed Columns` → `Added Column`

Each step in that chain does one thing:
- **Source** — connect to the data source
- **Navigation** — select the table or sheet
- **Remove Columns** — drop unneeded fields
- **Filter Rows** — remove out-of-scope data
- **Changed Type** — assign column data types
- **Renamed Columns** — apply naming conventions
- **Added Column** — row-level calculation

**Key Properties of Steps:**

| Property | What It Means |
|---|---|
| **Named** | Steps have descriptive names like 'Removed Columns' or 'Changed Type' that can be edited |
| **Editable** | Click any step in the Applied Steps pane to inspect or modify its settings or M code |
| **Ordered** | Steps run top-to-bottom — reordering or deleting any step changes all subsequent output |
| **Repeatable** | When source data refreshes, all steps re-run automatically — no manual repetition needed |

> Steps can also apply to multiple columns simultaneously — e.g. 'Changed Type' 
> can retype 5 columns in a single step.

### 3.A.3 Power Query Automates Manual Processes

Power Query replaces repetitive manual data preparation work with a recorded, 
refreshable pipeline. Think of it as a **macro language for data prep** — record 
steps once, run them repeatedly and automatically whenever source data changes.

| Without Power Query (Manual Excel) | With Power Query (Automated) |
|---|---|
| Open the source Excel file | Build query steps once using the GUI |
| Manually delete unnecessary columns | Columns removed automatically every refresh |
| Filter out rows outside scope | Filters applied automatically every refresh |
| Fix data types and blank values | Type & null fixes applied automatically |
| Copy cleaned data into reporting workbook | Model updated automatically in Power BI |
| Repeat everything from scratch when new data arrives | Click 'Refresh Data' — all steps re-run instantly |

### 3.A.4 Query Folding

Query folding is a performance optimisation that pushes transformation logic 
back to the data source — reducing the amount of data transferred into Power BI.

**Without query folding:** SQL Server sends ALL rows to Power BI, which then 
filters them locally — slow and wasteful.

**With query folding:** The filter runs as SQL on the server, and only matching 
rows are sent to Power BI — fast and efficient.

**What gets pushed to the source:**

- **Filtering** — WHERE clauses run on the DB, only matching rows are sent
- **Column Exclusion** — SELECT only needed columns, less data transferred
- **Joining** — JOINs happen at the source using native indexes
- **Grouping** — GROUP BY runs on the DB, pre-aggregated results returned

> Query folding is supported by SQL Server, MySQL, PostgreSQL, and other 
> relational databases. **Flat files and some APIs do not support it.**

### 3.A.5 Power Query Uses

Power Query is a powerful tool for self-service ETL — but knowing its limits 
helps you choose the right tool for the job.

**Power Query excels at:**
- Extracting data from a variety of sources (files, DBs, APIs, web)
- Shaping data — splitting, merging, pivoting, unpivoting tables
- Merging data — joining tables from different sources
- Transforming data — changing types, formats, and structures
- Cleaning data — removing nulls, fixing typos, standardising formats
- Calculating values — row-level formulas and conditional columns

**Limitations — when to consider alternatives:**
- Auditing or logging transformation activity
- Strict data type enforcement across sources
- Non-linear or branching data preparation logic
- Performance / scalability with large datasets
- Pipelines exceeding ~20 steps begin to degrade
- Complex enterprise ETL → consider **SQL Server Integration Services (SSIS)**

>  Power Query is optimised for ease-of-use over raw speed. For enterprise 
> ETL, Microsoft SQL Server Integration Services (SSIS) is a popular alternative.

### 3.A.6 Products that use Power Query

Power Query is not exclusive to Power BI — the same transformation engine is 
embedded across the Microsoft data platform. Skills learned here transfer directly.

| Product | How Power Query is Used |
|---|---|
| **Power BI Desktop** | Primary focus of this course — all transformations run before the semantic model loads |
| **Microsoft Excel** | Available as 'Get & Transform Data', also called Power Pivot when used for data modeling |
| **Power BI Service (Dataflows)** | Cloud-hosted Power Query — enables shared, centralised transformations reusable across reports |
| **Azure Data Factory** | Enterprise cloud ETL service — Power Query available as a data flow activity within pipelines |
| **Microsoft Dataverse** | Power Platform's data platform — Power Query maps and imports data from external sources |
| **SQL Server Analysis Services** | On-premises and Azure SSAS Tabular models share the same xVelocity engine and M language foundation |

>  Because the engine is shared across products, skills learned in Power BI 
> Desktop transfer directly to Excel, Azure Data Factory, and more.

### 3.A.7 Dataflows

Dataflows bring Power Query to the cloud — decoupling data transformation 
from the modeling and visualization layers.

**Power BI Desktop (Coupled)**  
All four layers are tightly coupled in one .pbix file:
`Data Source` → `Power Query (M)` → `Semantic Model` → `Report Visuals`

**Dataflows (Decoupled, Cloud)**  
The transformation layer lives in the Power BI Service and is shared:
- A **Dataflow** runs Power Query M on a schedule in the cloud
- Multiple Power BI Desktop files can consume the same dataflow as a data source
- Semantic model and visuals remain local in Desktop

> Dataflows are ideal for teams sharing the same cleaned dataset across 
> multiple reports — **transform once, use everywhere.**

### 3.A.8 Power Query Formula Language M

M is the language the Power Query Editor writes automatically as you use the 
GUI — and that you can edit directly for advanced tasks.

- **Auto-Generated by the GUI** — every click in Power Query Editor produces M 
  code behind the scenes. No coding required to build a functional query.
- **Inconsistent Naming** — Microsoft refers to it variously as "M", "Power Query M", 
  or "Power Query Formula Language". All the same thing.
- **Advanced Editor** — surfaces the full M script for a query. Start from 
  GUI-generated code and modify it to handle advanced scenarios.
- **Unlocks Advanced Tasks** — writing your own M enables parameterisation, 
  dynamic source paths, custom functions, and complex conditional logic.

**Example M script (auto-generated by the GUI)**:

let
Source = Csv.Document(File.Contents("data.csv")),

#"Promoted Headers" = Table.PromoteHeaders(Source),

#"Removed Columns" = Table.RemoveColumns(

#"Promoted Headers",

{"ID","Temp"}),

in

#"Removed Columns"

<h2 style="color:steelblue">Section 3.B: Power Query Editor Tour</h2>

### 3.B.1 Getting to the Power Query Editor

Two ways to open Power Query from Power BI Desktop:
- **Home ribbon → Get Data** — connect to and load a new data source
- **Home ribbon → Transform Data** — opens Power Query Editor for an existing connection

The **Data** and **Queries** sections of the Home ribbon contain all Power 
Query-related shortcuts.

---

### 3.B.2 Supported Data Sources

Power Query connects to a wide variety of sources:

| Category | Examples |
|---|---|
| **Databases** | SQL Server, MySQL, Oracle, PostgreSQL — local, on-premises, or cloud |
| **Files** | Excel, Text, CSV, XML, JSON — stored locally, in SharePoint, or OneDrive. Can import an entire folder at once. |
| **Web** | URL sources, OData feeds |
| **Microsoft Fabric & Azure** | Dataflows, Warehouses, Lakehouses, Cloud Databases, Blob Storage |
| **Online Services (SaaS)** | Salesforce, Google Analytics, Adobe, and others |
| **Power Platform** | Dataflows, Dataverse, Data Service (Legacy) |

### 3.B.3 Power Query Editor: Interface Tour

The Power Query Editor has four main areas:

**1. Ribbon Pane (top)**
- Contains shortcuts to bring in or transform data contextually
- Tabs include: Home, Transform, Add Column, View, Tools, Help

**2. Queries List Pane (left)**
- Lists all queries (tables) already in your project
- Queries can be organised into folders when you have many
- Click a query to select it and preview its data

**3. Query Settings Pane (right)**
- Lists the Applied Steps that make up the selected query
- Steps can be reordered or deleted here
- A ⚙️ gear icon appears next to some steps — click it to modify 
  the step using a graphical dialog

**4. Body / Data Preview (centre)**
- Shows a preview of the data after the currently selected step is applied
- Right-click columns to see a context menu
- Double-click a column header to rename it
- Drag column headers to reorder

**5. Expression Editor (formula bar)**
- Shows the M code behind the selected step
- Can be expanded from one line to multiple lines for easier editing

<h2 style="color:steelblue">Section 3.C: Simple Transformtion</h2>

### 3.C.1 Renaming a Query / Table

1. Open Power Query Editor
2. In the Queries List pane, find the query (table) you want to rename
3. **Right-click** on it → select **Rename**
4. Type the new name and press Enter

> You can also double-click the query name directly in the Queries pane, 
> or press **F2** to make it editable.

### 3.C.3 Renaming Columns

1. Open Power Query Editor
2. Select the query that contains the column you want to rename
3. Click on the column header to select it
4. **Right-click** → select **Rename**, or simply **double-click** the column header
5. Type the new name and press Enter

### 3.C.4 & 3.C.5 Changing a Column Data Type

Selecting the right data type is critical — it determines which transformations 
and aggregations are available for that column.

- **Numeric types** (Decimal, Whole Number) — enable Sum, Average, and other numeric aggregations
- **Text type** — enables transformations like Split Column; disables numeric aggregations
- **Date type** — enables time intelligence functions and date hierarchies
- Power BI sometimes **hides or deactivates** transformations that don't apply 
  to the current data type — this is a clue that your type may be wrong

**Know your source:**
- **SQL Server** — carries data type info over; usually correct, but still worth reviewing
- **Text files (CSV, TSV)** — Power Query guesses data types — always review column by column!

**How to change a data type:**
1. In Power Query Editor, select the query
2. Click the column you want to retype
3. Go to the **Home ribbon** → in the Transform section, click the **Data Type** dropdown
4. Select the correct type

### 3.C.6 Adding Calculated Fields (Custom Columns)

A calculated field adds a new column whose values are derived from existing 
columns using an M expression.

**How to add one:**
1. In Power Query Editor, select the query
2. Go to the **Add Column** tab in the ribbon
3. Click **Custom Column**
4. Give the new column a name
5. Write the M expression in the formula box using `[ColumnName]` syntax

**Example — combining name fields:**

Employee Title of Courtesy] & " " & [Employee First Name] &
    
" " & [Employee Last Name] & " (" & [Employee Title] & ")"

This produces a single `Employee Description` column like:
`"Ms. Nancy Davolio (Sales Representative)"`

> The formula box shows available columns on the right — click 
> **Insert** to add them rather than typing manually to avoid typos.
> Power Query will flag syntax errors before you click OK.

## 🗝️ Module 3 — Key Takeaways

**Power Query is your data prep workhorse.**  
It sits between your raw data sources and your semantic model — shaping, 
cleaning, and structuring everything before Power BI sees it.

**Every action becomes a step — and that's a superpower.**  
Steps are named, ordered, editable, and repeatable. When your source data 
updates, just hit Refresh and every step re-runs automatically.

**Query folding makes things faster without extra effort.**  
When your source supports it (SQL Server, MySQL, etc.), Power Query pushes 
filtering and joins back to the database — less data travels, faster results.

**Know the limits.**  
Power Query is great for self-service ETL up to ~20 steps. Beyond that, 
or for enterprise-scale pipelines, consider SQL Server Integration Services (SSIS).

**The GUI writes M for you — but you can take over when needed.**  
Every click generates M code behind the scenes. For advanced scenarios, 
open the Advanced Editor and modify or write M directly.

**Power Query skills transfer across the Microsoft ecosystem.**  
The same engine runs in Excel, Azure Data Factory, Dataverse, and more. 
Learn it once, use it everywhere.

**The four transformations you must know cold:**  
Rename a table → Remove columns → Rename columns → Change data types → Add a calculated field.  
These are the building blocks of every real-world Power Query project.
