### Week 9 | June 5, 2026

**Visualizations** are the final layer of the Power BI pipeline — where your 
cleaned data model and DAX calculations become charts, tables, maps, and 
dashboards that people can read, interact with, and act on.

the right chart type, configure visuals correctly, ensure your data supports

## The Power BI Report Creation Workflow

Before diving into visuals, it helps to see how Module 5 fits into the 
full end-to-end pipeline:

**Data Sources → Load → Power Query (M) → Data Modeling → DAX → Visuals → Interactive Report**

| Stage | Tool | What Happens |
|---|---|---|
| **1. Data Sources** | Get Data | Connect to Excel, SQL, APIs, SharePoint, Cloud SaaS |
| **2. Load** | Import / DirectQuery | Load data into model; schema detection; query folding |
| **3. Power Query (M)** | Power Query Editor | Remove duplicates, filter rows, rename columns, change types, merge/append, add custom columns |
| **4. Data Modeling** | Model View | Star schema design, relationships, hierarchies, calculated columns, RLS |
| **5. DAX** | Formula Bar | Measures (SUM, CALCULATE), calculated columns, time intelligence |
| **6. Visuals** | Report Canvas | Bar, line, pie, scatter, map, table, KPI, card — this module |
| **7. Interactive Report** | Report View | Slicers, drill-through, cross-filtering, bookmarks, publish & share |

> Visualization is not the first step — it's the last. A great visual 
> built on bad data or a poorly structured model will always mislead.

<h2 style="color:steelblue">Section 5.A: Working with Visualizations</h2>

### 5.A.1 Choosing Visuals

The shape of your visualization should support the story — start with the 
message, then choose the chart that delivers it most clearly.

**Support the Story**
- Consider the level of detail needed on the page — too much detail overwhelms
- Think about what interactions you want your end user to have (drill-down, filter, compare)
- Match the chart type to the nature of your data (categorical, time-series, part-of-whole)

**Prioritise Clarity**
- How will your audience access the report? (desktop, mobile, projected screen)
- What context do they need to understand the visuals?
- Avoid clutter — fewer, clearer visuals beat many busy ones

**Consider Your Audience**
- **Executives** want summary KPIs, not raw tables — give them cards and gauges
- **Analysts** need drill-down and filter capability — give them matrices and scatter charts
- Report layout should guide the eye to key insights first

**Accessibility**
- Assume users view on a smaller screen than yours — design for that
- Use accessible colour schemes — check the View tab for built-in accessible themes
- Tools to verify colour accessibility: 
  - DaltonLens (daltonlens.org) — desktop colour blind checker
  - Adobe Color Blind Safe (color.adobe.com) — online accessibility checker
- Add labels, icons, and different line styles (dotted, dashed) to augment colour meaning
  - Labels improve screen-reader compatibility
  - Icons and textures support grayscale printing

### 5.A.2 Configuring Visuals

### 5.A.2 Configuring Visuals

Every visual has **field wells** — slots where you drag fields to define what 
the visual shows. Available wells vary by chart type:

| Field Well | Typical Data Type | Example |
|---|---|---|
| **X-Axis** | Categorical / qualitative | Product, Region, Date |
| **Y-Axis / Values** | Numeric / quantitative | Sales, Count, Revenue |
| **Legend** | Categorical / qualitative | Category, Status, Gender |
| **Small Multiples** | Categorical | Creates one chart per value — e.g. one chart per Region side-by-side |
| **Tooltip** | Any | Extra detail available on hover without cluttering the visual |

**Format checklist — ask yourself before finalising any visual:**
- [ ] Are colour choices unambiguous? (check with accessibility tools)
- [ ] Are label meanings clear and readable?
- [ ] Is the format of the visual useful for this data?
- [ ] Is the order and arrangement of visuals logical on the page?

### 5.A.3 Visuals Need Good Data

It's very common to discover data issues once you start creating visualizations 
— treat that as a normal part of the process, not a failure.

**Model Shape Matters**
- Additional measures and calculated columns may be needed to support your visuals
- Relationships may need adjustment — e.g. a visual that needs data from two tables 
  requires a working relationship between them

**Performance Considerations**
- Large datasets + complex visuals = slower load times
- Be thoughtful about the number and type of visuals per page
- Avoid unnecessary calculated columns that bloat model size

**Expect to Find Data Issues**
- Missing values, wrong data types, or incorrectly named columns often only 
  surface once you drag fields into a chart

**Iterate Between Layers**
- Visualization ↔ DAX ↔ Power Query ↔ Model are not one-way
- Be ready to loop back to any layer to fix what you see in the report
- Example: you build a line chart by month but months sort alphabetically → 
  go back to the model and set Sort By Column (Month Name → Month Number)

### 5.A.4 Types of Visualizations

Power BI offers a rich library of visuals — organised by what analytical 
question they answer best:

**Aggregate & Compare**
Bar chart · Column chart · Line chart · Area chart · Scatter chart · Bubble chart

**Highlight a Data Point or Goal**
Card · Gauge chart · KPI visual · Scorecard

**Part of a Whole**
Pie chart · Donut chart · Treemap · Waterfall chart

**Detailed Data**
Table · Matrix

**Non-Data Visual Elements**
Text box · Image · Button (forward/back) · Link · Shape

> Power BI also supports **custom visuals** from the AppSource marketplace 
> — community and vendor-built visuals that extend the default library.

### 5.A.5 Bar Charts and Column Charts

Simple and powerful visuals for comparing values across categories. 
Sub-values in a category can be stacked or clustered.

**Bar Chart (Horizontal)**
- Best when: category labels are long · many categories to compare
- Use for: ranking, many categories
- Example: Total Invoice by Tier — Tier 1 ($188M), Tier 2 ($20M), Tier 3 ($7M)...
  → horizontal bars make it easy to read long labels without rotation

**Column Chart (Vertical)**
- Best when: time series · fewer categories
- Use for: time trends, category comparison
- Example: Total Invoice by Sub Category — bars go up across months or categories

**Sub-value options:**
- **Stacked** — bars extend to show each segment's contribution to a total
  - Example: one bar per month, coloured segments showing Direct vs Indirect vs Logistics
- **Clustered** — bars sit side by side for direct comparison
  - Example: two bars per month — one for 2024, one for 2025

**100% Stacked variant:**
- Both bar and column charts support a 100% stacked version
- Each bar fills the full height/width to show relative proportions only
- Useful when the absolute total doesn't matter, only the breakdown does

> 💡  When in doubt between bar and column, ask: are you comparing over time 
> (column) or comparing many categories with long names (bar)?

### 5.A.6 Line Charts and Area Charts

Line charts trace a series of values over time to reveal trends. Area charts 
fill the shape to emphasise the magnitude of change.

**Line Chart**
- Connects data points with a line to show direction and trend
- Best for: time series with many data points, or when gaps between points matter
- Example: Total Invoice by Month and Category — four coloured lines (Direct, Indirect, 
  Logistics, Other) tracked across Jan–Dec showing how each category trended

**Area Chart**
- Same as a line chart but the area beneath the line is filled
- Emphasises the magnitude and volume of values, not just direction
- Best for: comparing cumulative magnitudes across series
- Example: same data as above but filled — makes it visually clear which category 
  has the largest volume at any given time

> 💡 Use a line chart when the number of data points is high or the gaps between 
> them matter. Use area when you want to compare cumulative magnitudes.

### 5.A.7 Pie Charts, Donut Charts, and Treemaps

All three show parts within a whole. The whole should add up to 100% — 
no overlapping segments.

**Pie Chart**
- Divided circle where each slice = a category's share of the total
- Slices are sized proportionally to their value
- Best for: simple part-of-whole with 5 or fewer categories

**Donut Chart**
- Same as a pie chart but with a blank hole in the middle
- The centre can display a summary metric (e.g. total count, grand total)
- Example: Total Revenue with 36 items shown in the centre hole
- Slightly easier to read than pie for comparing adjacent slices

**Treemap**
- Displays relative values as rectangles within a rectangle
- Largest values start top-left, arranged down to smallest at bottom-right
- Colour can encode a second dimension (e.g. category)
- Example: Hardware ($65M) top-left, Indirect ($55M) top-right, 
  Other ($42M), Raw ($26M), Out. ($22M), Logistics ($18M)
- Best for: when you want to show hierarchy and proportions simultaneously

> ⚠️ **Avoid pie/donut charts with more than 5–6 slices.** Small slices 
> are hard to compare. Use a bar chart instead when you have many categories.

### 5.A.8 Scatter and Bubble Charts

Scatter charts plot data points against two numerical axes — ideal for 
revealing clusters, correlations, and outliers.

**Scatter Chart — Two Numeric Axes**
- Both X and Y axes are numeric — unlike bar/column where X is categorical
- Each dot = one data point (e.g. one city, one product, one customer)
- Best for: correlation analysis, outlier detection, cluster identification
- Example: X = Total Invoice, Y = Discount % by City — reveals which cities 
  have high invoices AND high discounts (potential concern)

**Bubble Chart — Adds a Size Dimension**
- Same as scatter but bubble SIZE encodes a third measure
- Gives three dimensions of data in one chart: X, Y, and size
- Example: X = Total Invoice, Y = Discount %, Size = Discount Savings by City
  → you can now see which cities have the largest absolute discount savings

| | **Scatter Chart** | **Bubble Chart** |
|---|---|---|
| **Axes** | X and Y both numeric | X and Y both numeric |
| **3rd dimension** | None | Size of bubble |
| **Best for** | Clusters, trends, outliers | Three-variable relationships |

> 💡 Use scatter/bubble charts when both axes represent continuous measurements 
> — not categories. They're powerful for exploratory data analysis.

### 5.A.9 Table and Matrix

Tables and matrices display detailed data in rows and columns — providing 
precision that charts cannot.

**Table — Detailed Lookup**
- Shows every row of data, column by column
- Best for: quickly seeing and comparing detailed data by category row-by-row
- Example:

| Product | Category | Units | Revenue | Margin |
|---|---|---|---|---|
| Widget Pro | Electronics | 1,240 | $62,000 | 38% |
| Gadget X | Electronics | 856 | $42,800 | 42% |
| Cable Set | Accessories | 3,100 | $15,500 | 61% |
| **Total** | | **7,426** | **$236,850** | **40%** |

**Matrix — Breakdown with Subtotals**
- Shows data across multiple dimensions with subtotals and grand totals
- Like a pivot table — rows AND columns are both categories
- Best for: cross-tabular analysis, sub-region breakdowns, quarterly comparisons
- Example: Region (East/West) × Quarter (Q1–Q4) with subtotals per region 
  and a grand total row

| Region | Q1 | Q2 | Q3 | Q4 | Total |
|---|---|---|---|---|---|
| East | $2.1M | $2.4M | $2.8M | $3.1M | $10.4M |
| West | $1.8M | $1.9M | $2.1M | $2.5M | $8.3M |
| **Grand Total** | $3.9M | $4.3M | $4.9M | $5.6M | **$18.7M** |

> 💡 Tables show every row. Matrices summarise at multiple levels — 
> region → sub-region → product. Use conditional formatting to highlight anomalies.

### 5.A.10 & 5.A.11 Maps and Cards

**Basic Maps (5.A.10)**

Map visuals associate both categorical and quantitative data with a location — 
making geographic patterns instantly visible.

- Plot individual data points or size bubbles on a value dimension
- Requires geographic fields with the correct **Data Category** set:
  City, Country, State/Province, Postal Code, Latitude, Longitude
- Example: Sales by City — bubble size = total sales value
  - LA ($14M), NYC ($18M), DEN ($7M), CHI ($4M), DC ($9M), MIA ($5M)
  - Instantly shows NYC and LA as the top two markets

> ⚠️ If your map doesn't render correctly, check that the column's 
> Data Category is set to the right geography type in Table View.

---

**Cards (5.A.11)**

Card visuals display a single large KPI value — the simplest and most 
impactful visual for an executive summary.

**Three card uses:**
- **Single data point** — one large KPI value: `Total Revenue: $236.8M`
- **Multiple data points** — a row of KPI values side-by-side:
  - Total Revenue: $236.8M (+12%) · # of Orders: 14,726 (+8%) · 
    Avg Order Value: $1,609 (-3%) · Gross Margin: 40.2% (+1.4pp)
- **Reference card** — display a hard-coded label or date:
  - `Report Last Updated: June 5, 2026`

> 💡 Cards are typically placed at the top of a report page to give 
> executives an instant summary before they explore the detailed visuals below.

### 5.A.12 Scorecards, Gauges & KPI Visuals

These visuals communicate progress toward a goal — turning raw numbers 
into a performance story.

**Gauge Chart**
- Displays progress toward a goal as a **shaded arc**
- A target line marks the goal
- Example: 68% of $1M goal → arc fills to the 68% mark, target line at $1M
- Best for: showing how far ahead or behind you are on a single metric

**KPI Visual**
- Shows actual value vs. target + direction of trend in one compact visual
- Example: Sales Total $687K · Target: $1.0M · ▼ 31% below target
- Includes a trend sparkline showing direction even without a full chart

**When to use these visuals:**

- **"What are we ahead / behind on?"** → KPI visual shows actual vs. target + trend direction
- **"How far is this ahead / behind?"** → Gauge shows distance to target as a filled arc
- **"Where do we stand at a glance?"** → Cards + gauges on a summary page give quick answers
- **"How is the trend moving?"** → KPI trend line shows direction even without a full chart

> ⚠️ Use gauges and KPIs sparingly — they work best as headline indicators 
> on executive dashboards, not as the primary analytical tool.

### 5.A.13 Non-Data Visual Elements

Images, text boxes, forward/back buttons, links, and other visual elements 
can be added to a report from the **Insert tab** in Report View.

Uses:
- **Text boxes** — add titles, section headers, explanatory notes
- **Images** — incorporate brand logos, icons, or contextual images
- **Buttons (forward/back)** — create navigation between report pages
- **Links** — direct users to external resources or related reports
- **Shapes** — group visuals visually, create backgrounds, add dividers

> 💡 Non-data elements make reports feel like polished, professional 
> products — not just a collection of charts. They improve navigation, 
> reinforce brand identity, and guide the reader's eye.

## 🗝️ Module 5 — Key Takeaways

**Start with the message, then choose the chart.**
Never pick a chart type first. Ask: what am I trying to show? Comparison? 
Trend? Part of whole? Then choose the visual that delivers that story most clearly.

**Know your chart types cold:**
- 📊 **Bar / Column** — compare values across categories (bar = many/long labels, column = time)
- 📈 **Line / Area** — trends over time (line = direction, area = magnitude)
- 🥧 **Pie / Donut / Treemap** — parts of a whole (max 5–6 slices for pie/donut)
- 🔵 **Scatter / Bubble** — two numeric axes, correlations, outliers (bubble adds a 3rd dimension)
- 📋 **Table** — detailed row-level lookup
- 🔢 **Matrix** — cross-tabular breakdown with subtotals (like a pivot table)
- 🗺️ **Map** — geographic distribution (requires correct Data Category on geographic fields)
- 🃏 **Card** — single KPI value at a glance, best at top of page
- 🎯 **Gauge / KPI** — progress toward a goal, use sparingly as headline indicators

**Field wells determine what a visual shows.**
X-axis = categorical, Y-axis = numeric, Legend = categorical, Tooltip = any. 
If a visual looks wrong, check what's in each well first.

**Good visuals require good data.**
It's normal to discover data issues when you start building visuals. 
Loop back to Power Query, the model, or DAX as needed — it's all connected.

**Design for your audience, not yourself.**
Executives need summary KPIs. Analysts need drill-down. 
Design for a smaller screen than you're using. Use accessible colour schemes.

**Non-data elements make the difference between a chart dump and a report.**
Titles, buttons, images, and shapes turn a collection of visuals into 
a polished, navigable product.
