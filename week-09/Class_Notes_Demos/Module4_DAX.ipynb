<h2 style="color:steelblue">Section 4.A: Defining DAX</h2>

### 4.A.1 Data Analysis eXpression Language

**DAX** stands for **D**ata **A**nalysis e**X**pression Language.

- Developed by Microsoft as the expression language native to the 
  **in-memory columnar database** (xVelocity) used for information analysis
- Used to add meaning to your data through formulas and expressions
- Used everywhere xVelocity is used — Power BI, Power Pivot, SSAS Tabular
- Adds four types of calculations to your model:
  - **Calculated Columns** — row-by-row values stored in the model
  - **Measures** — saved calculations that run on demand
  - **Calculated Tables** — new tables created via DAX expressions
  - **Row Filters** — controls which rows a user can see (row-level security)

> Learning DAX is essential to getting the most out of Power BI — 
> it's what turns a data model into an analytical tool.

### 4.A.2 DAX and the Data Model

How your data model is shaped will drastically affect how useful your DAX 
analytics can be. **A good model comes first — then DAX.**

DAX is used both to help model data AND to analyze modeled data. It creates 
new information from data already in your model. Examples:
- **Percentages** — what % of total sales does this region represent?
- **Summaries** — total revenue, order counts, average order value
- **KPIs** — Key Performance Indicators like Sales vs. Target
- **Trends** — how is this metric changing over time?

DAX also helps manage **granularity** — the level of detail in your data:
- It can turn an order-level discount into a line-item discount for 
  reports that work at the order detail row level
- It can summarize data with totals or percentages *after* that data 
  has been filtered to a specific subset

> Invest in proper modeling before writing complex DAX — 
> DAX is only as powerful as the model it works with.

### 4.A.3 Analysis with DAX Calculations

DAX is most useful for two types of analysis:

**Aggregation** — summarising many detail records into something meaningful
- When you have millions of rows, you need a way to see the data without 
  manually interpreting each number
- Example: instead of seeing 50,000 individual sales rows, DAX gives you 
  "Total Sales = $1,354,458" broken down by category in a chart

**Filtering** — slicing data to a specific subset
- Example: "Total Sales in the Northeast Region in Q3" filters before aggregating

Built-in DAX function categories that help with analysis:
- **Statistical Analysis** — SUM, AVERAGE, MIN, MAX, STDEV
- **Time-based Analysis** — year-to-date, month-over-month, same period last year
- **Dimension slicing** — grouping and filtering through user-defined hierarchies

### 4.A.4 Calculations Not Suitable for Storage

Some calculations *cannot* be pre-stored in the dataset because their result 
depends on a context that changes at report time. These are exactly what 
**DAX measures** are designed for.

A calculation is not suitable for storage when:
- The result changes if data in another row changes
- The calculation needs to be performed across many different filter contexts

**Real examples:**
- *"For a given line item, what % of that order does this single line represent?"*  
  → This changes depending on which order you're looking at — can't be stored as a fixed number
- *"What was the average customer age at time of sale?"*  
  → Changes depending on which time period or region is being filtered
- *"Correctly calculated percentages at different grouping levels"*  
  → A % of total changes depending on what "total" means in the current filter context
- *"Aggregate summaries"* — totals that shift as users filter the report

> Think of it this way: a stored column has one fixed value per row. 
> A measure has no stored value — it calculates fresh every time the 
> report filter context changes.

### 4.A.5 DAX is an Expression Language

DAX is a **collection of functions, operators, and constants** that can be 
used in a formula or expression to calculate and return one or more values.

**DAX vs Excel — the key difference:**

| | Excel | DAX |
|---|---|---|
| **References data by** | Cell coordinates (e.g. `A1`, `B2:B10`) | Named identifiers — table and column names |
| **Knows relationships?** | No — you manage lookups manually | Yes — understands the data model relationships |
| **Scope** | Works on a spreadsheet grid | Works across a relational semantic model |

Writing DAX feels similar to Excel formulas, but instead of `=SUM(B2:B10)` 
you write `SUM(Sales[Sales Amount])` — referencing the table and column by name.

>  Because DAX understands relationships between tables, it can reach across 
> your entire model to calculate things Excel never could.

### 4.A.6 – 4.A.9 DAX vs Power Query

**What they share:**
- Both can calculate column values
- Both can calculate entire tables
- Both can rename columns and tables
- Either can feel more intuitive depending on the situation

**Where they differ:**

| | **Power Query** | **DAX** |
|---|---|---|
| **When it runs** | Before data enters the model | After data is in the model |
| **How it works** | Step-by-step pipeline | Single nested expression |
| **Feels like** | SQL / ETL workflow | Spreadsheet formula language |
| **Query Folding** | Yes — can push work to the source DB | No |
| **Dynamic measures** | ❌ Cannot create filter-context measures | ✅ Core strength |
| **Cross-table calculations** | Limited | ✅ Understands relationships |

**When to use Power Query:**
- Bringing data into the model
- Dropping unnecessary columns to keep the model lean
- Reusing query patterns across projects
- Any structural shaping before modeling begins

**When to use DAX:**
- Creating measures that recalculate based on filter context
- Building complex calculations that span across multiple tables
- Creating KPIs, percentages, and time-based comparisons
- Any calculation that needs to be *dynamic* at report time

> Use both together — Power Query shapes and loads, DAX analyses and enriches.

### 4.A.10 DAX Family of Products

DAX knowledge is transferable — the same language runs across multiple 
Microsoft products:

- **Power Pivot for Excel** — available as an add-on since 2010, built-in since Office 2016
- **Power BI Desktop** — local authoring (this course)
- **Power BI Service** — cloud publishing and collaboration
- **Power BI Report Server** — on-premises Power BI for organisations
- **SQL Server Analysis Services (SSAS) Tabular** — enterprise-scale analytical models, available since 2012

> Learn DAX once in Power BI Desktop and your skills apply across the 
> entire Microsoft BI ecosystem.

<h2 style="color:steelblue">Section 4.B: Getting Started with DAX</h2>

### 4.B.1 & 4.B.2 Implicit vs Explicit DAX Calculations

**Implicit measures** are created automatically by Power BI behind the scenes 
when you drag a numeric field into a visualization's data well.

How it works step by step:
1. You drag a numeric field (e.g. `[Amount]` from `'order detail'`) to the Y-axis
2. Power BI silently creates a DAX measure behind the scenes: `SUM('order detail'[Amount])`
3. The Y-axis label becomes: **"Sum of Amount"**
4. You can modify the aggregation type by clicking the field in the data well 
   and choosing: Sum · Average · Minimum · Maximum · Count (Distinct) · Count · Standard Deviation

⚠️ With an implicit measure the formula is hidden — you can only switch 
aggregation type by clicking. **You cannot reuse it in other calculations.**

---

**Explicit measures** are named DAX calculations you write yourself — they 
appear in the model and can be reused anywhere.

Why create explicit measures?
- Assign a clear, meaningful business name (e.g. `Sales Total` instead of `Sum of Amount`)
- Reuse in other calculations — one measure can reference another as a building block
- Control and modify — change the DAX formula directly instead of relying on a dropdown
- Can range from simple to complex — from a single SUM to multi-table calculations with filter context

**Implicit vs. Explicit — side by side:**

| | **Implicit (auto-created)** | **Explicit (you create it)** |
|---|---|---|
| **Formula** | `SUM('order detail'[Amount])` hidden | `Sales Total = SUM('order detail'[Amount])` visible |
| **Y-axis label** | "Sum of Amount" | "Sales Total" |
| **Named?** | No | Yes |
| **Reusable?** | No | Yes — can be used inside other measures |
| **Change aggregation** | Click to switch type | Edit the DAX formula directly |

> An explicit measure named `[Sales Total]` can be used inside another measure:
> `Avg Order Value = [Sales Total] / [# of Orders]`

### 4.B.3 Writing DAX Expressions

DAX expressions are written in the **formula bar** — which appears when 
creating a new Measure, Calculated Column, or Calculated Table.

**Expression Anatomy:**

`Sales Total = SUM('order detail'[Amount])`

Breaking that down:
- `Sales Total` → the **Name** you give the calculation
- `=` → separates the name from the expression
- `SUM()` → the **Function** being used
- `'order detail'` → the **Table** name (single quotes because it has a space)
- `[Amount]` → the **Column** name (always in square brackets)

**Key Syntax Rules:**

**1. Name = Expression** — every DAX calculation starts with a name, then `=`, then the formula

**2. Table names with spaces → single quotes**
- Has a space: `SUM('order detail'[Amount])`
- No space needed: `SUM(Orders[Amount])`

**3. Field names always in square brackets**
- Column names, measure names — always wrapped in `[ ]`
- Examples: `[Sales Total]` · `[Amount]` · `[# of Orders]`

**4. Save with ✅ or Enter**
- Click the green checkmark in the formula bar, or press Enter to save
- The red ❌ X cancels without savingess the **green check ✅** or hit **Enter** to save the expression to the model

**Example expressions:**
```
-- Referencing a column (table name required)
Sales Total = SUM(Sales[Sales Amount])

-- Referencing another measure (no table name needed)
Avg Order Value = [Sales Total] / [# of Orders]

-- Hard-coded value
Annual Sales Goal = 1000000

-- Derived from another measure
Monthly Sales Goal = [Annual Sales Goal] / 12
```

### 4.B.4 What is a Measure?

A **measure** is a named DAX calculation saved to the model — it runs only 
when called, and can be used by other measures.

**Four key properties:**
- **Saved to the Model** — appears in the Fields pane, available in every report built on the model
- **Runs On Demand** — only evaluated when placed in a visual or used by another measure
- **Reusable as a Variable** — one measure can reference another, building complex logic in layers
- **Context-Aware** — automatically recalculates for each filter context: product, region, date range, etc.

**Measure examples — all based on `'order detail'` table:**

`Sales Total = SUM('order detail'[Amount])` — adds up all sales amounts

`# of Orders = COUNTROWS('order detail')` — counts total number of rows

`Avg Order Value = [Sales Total] / [# of Orders]` — measure referencing two other measures

`Annual Sales Goal = 1000000` — hard-coded value saved as a measure

`Monthly Sales Goal = [Annual Sales Goal] / 12` — derived from another measure

> 💡 Build measures in layers — write and test simple ones first, then 
> combine them into more complex expressions.


### 4.B.5 What is a Calculated Column?

A **calculated column** uses DAX to add a new column to a table — values 
are computed row-by-row and **stored permanently in the model**.

**Key properties:**
- **Computed at model refresh** — runs once when created, then re-runs every time data refreshes
- **Values stored in the model** — takes up memory; each row gets its own value
- **Used like any regular column** — available for filtering, slicing, grouping, and in other DAX
- **Row context** — the DAX expression accesses each row's values directly, like a spreadsheet formula

**Example — Sales Tax column:**

`Sales Tax (7%) = 'order detail'[Amount] * .07`

Resulting table:

| Row | Amount | Sales Tax (7%) |
|---|---|---|
| 1 | $45.00 | $3.15 |
| 2 | $89.99 | $6.30 |
| 3 | $25.00 | $1.75 |
| 4 | $120.00 | $8.40 |

**Why `SUM()` wouldn't work here:**
A calculated column cycles through each row and operates on that row's `[Amount]` 
value individually. No aggregation is needed — and none would be valid. `SUM()` 
is for measures that need to summarise across many rows.

> ⚠️ Calculated columns add storage cost — use measures where the calculation 
> can be evaluated dynamically per visual.

---

### Measure vs. Calculated Column — Full Comparison

| | **Measure** | **Calculated Column** |
|---|---|---|
| **Created by** | User writes DAX in formula bar | User writes DAX in formula bar |
| **When computed** | At query time — when placed in a visual | At model load / refresh time |
| **Storage** | No storage — values never saved | Stored in model; increases memory use |
| **Context** | Filter context — adapts to slicers/filters | Row context — one value per row |
| **Can reference** | Other measures and columns | Columns in the same table row |
| **Best for** | Aggregations, KPIs, dynamic analysis | Row-level derivations (tax, full name, age bracket) |
| **Aggregation fn?** | Required (SUM, COUNT, AVERAGE...) | Not needed — works row by row |
| **Appears in** | Fields pane (calculator icon 🧮) | Table in Fields pane (like any column) |

> 💡 **Rule of thumb:** If the value can be calculated dynamically at report 
> time, use a **measure**. If each row genuinely needs its own stored value, 
> use a **calculated column**.

### 4.B.6 Common Aggregation Functions in DAX

These functions summarise column data — the building blocks of most DAX 
measures in real-world Power BI reports:

| Function | What It Does | Syntax | Example |
|---|---|---|---|
| `SUM()` | Total of all numeric values in the column | `SUM(Table[Column])` | `Sales Total = SUM('order detail'[Amount])` |
| `AVERAGE()` | Mean of all numeric values in the column | `AVERAGE(Table[Column])` | `Avg Amount = AVERAGE('order detail'[Amount])` |
| `COUNT()` | Number of non-blank numeric values | `COUNT(Table[Column])` | `# Values = COUNT('order detail'[Amount])` |
| `COUNTA()` | Number of non-blank values (any type including text) | `COUNTA(Table[Column])` | `# Non-Blank = COUNTA('order detail'[CustomerID])` |
| `COUNTROWS()` | Number of rows in the table or filtered table | `COUNTROWS(Table)` | `# of Orders = COUNTROWS('order detail')` |
| `MIN()` | Smallest value in the column | `MIN(Table[Column])` | `Earliest Date = MIN(Orders[OrderDate])` |
| `MAX()` | Largest value in the column | `MAX(Table[Column])` | `Latest Date = MAX(Orders[OrderDate])` |

**More DAX functions coming in Week 10:**
- `FILTER()` — filter a table before aggregating
- `CALCULATE()` — change the filter context of a calculation
- `ALL()` — remove all filters from a column or table
- `RELATED()` — pull a value from a related table across a relationship
- `USERELATIONSHIP()` — activate an inactive relationship for a specific calculation
- **Time Intelligence** — `DATEADD()`, `TOTALYTD()`, same period last year, etc.

## 🗝️ Module 4 — Key Takeaways

**DAX is the language that makes your data model think.**  
Power Query brings the data in and shapes it. DAX asks questions of it — 
totals, percentages, KPIs, trends, and anything that needs to recalculate 
dynamically as users interact with the report.

**Your model quality determines your DAX quality.**  
A messy, poorly structured model makes DAX painful to write. 
Invest in a clean Star/Snowflake schema first — then DAX becomes straightforward.

**DAX uses table and column names, not cell coordinates.**  
`SUM(Sales[Sales Amount])` not `SUM(B2:B100)`. This means DAX understands 
your relationships and can calculate across tables — something Excel can't do.

**Some things can't be stored — that's what measures are for.**  
Percentages, filter-context totals, and cross-table summaries change depending 
on what's filtered in the report. Measures recalculate live — calculated columns don't.

**Implicit measures are automatic, explicit measures are yours.**  
Drag a number onto a chart → Power BI creates an implicit measure silently.  
Write your own and name it → that's an explicit measure you can reuse everywhere.

**Measure vs Calculated Column — know the difference:**  
- 🧮 **Measure** — runs on demand, no stored values, recalculates with every filter change  
- 📋 **Calculated Column** — runs at refresh, stores one value per row, used for grouping and labeling

**Start with the basics, the rest comes in Week 10.**  
`SUM`, `COUNT`, `COUNTROWS`, `AVERAGE`, `MIN`, `MAX` — master these first.  
`CALCULATE`, `FILTER`, time intelligence, and `RELATED` are coming next week.
