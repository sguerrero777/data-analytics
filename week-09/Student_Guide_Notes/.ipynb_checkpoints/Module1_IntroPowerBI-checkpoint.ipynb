<h2 style="color:steelblue">Section 1.A: Getting Started with Power BI</h2>

### 1.A.1 What is Power BI?

`Power BI` is Microsoft's end-to-end **business intelligence solution**

It manages the full data pipeline in one tool:

  **Ingest → Clean → Transform → Model → Enhance → Analyze → Visualize**

- **I**: ngestion of data from a wide variety of sources
- **D**: ata cleaning
- **T**: ransformation and reshaping of data
- **D**: ata modeling
- **D**: ata enhancement
- **A**: nalysis & Visualization

### 1.A.2 Power BI Ecosystem

| Component | What It Does |
|---|---|
| **Power BI Desktop** | Free Windows app — author models & reports, save as `.pbix` |
| **Power BI Service** | Cloud/SaaS — publish, share, and collaborate on reports |
| **Power Query Editor** | Embedded transformation tool inside Desktop & Service |
| **Power BI Mobile** | iOS/Android/Windows — view published reports on the go |

> Power BI also plugs into the broader Microsoft ecosystem:
> Azure, Excel, SQL Server, SharePoint, Teams, etc.

### 1.A.3 Power BI Desktop

- Free, powerful **Windows-only** desktop app (not available on Mac)
- Projects saved as **.pbix files** — data stored in-memory via xVelocity
- Three important limitations to know:
  1. Data goes **stale** — must be manually refreshed
  2. **Doesn't save mulitple file versions automatically** → use GitHub to version .pbix files
  3. **Credential sharing issues** — other users may lack access to your data sources

### 1.A.4 Power Query and M Language

- The **data ingestion and transformation layer** of Power BI
- Fully embedded inside Power BI Desktop
- GUI-driven — but writes **M language** code behind the scenes to clean and transform data
- Think of it as: *get data ready and shaped BEFORE it enters the model*

### 1.A.5 Power BI Mobile

- Read-only consumption of reports published to the **Power BI Service**
- Available on Windows, iOS, and Android
- Connects to Power BI Service or Power BI Report Server


### 1.A.6 xVelocity Analytical Engine

- The **in-memory columnar database engine** that powers everything
- Also called: **VertiPaq** (codename) | **Power Pivot** (when used in Excel)
- `DAX` is an expression langauge that was written *specifically* for xVelocity
- Used across multiple Microsoft products:
  - Power BI Desktop → hosts model locally on desktop
  - Power BI Service → hosts model in the cloud
  - SQL Server Analysis Services (SSAS) Tabular Engine

### 1.A.7 Columnar Databases

**Columnar databases are ideal for analytical reporting**


| Row-Based (Traditional) | Columnar |
|---|---|
| Reads entire rows including unneeded columns | Reads only the columns requested |
| Great for transactions (INSERT/UPDATE) | Great for analytics (SUM, COUNT across millions of rows) |
| Wastes I/O on irrelevant data | Efficient — only loads what's needed |

> Analogy: Imagine a spreadsheet where each column lives in its own file.
> To compute Total Sales, you only open the Sales column — not every row of every field.

Computers read disk in `~64k byte chunks`. Columnar storage means those chunks
contain only the relevant column data, not unneeded fields from every row.

### 1.A.8 Data Analysis eXpression Language

- `DAX` stands **D**ata **A**nalysis e**X**pression Language
- Microsoft's native formula language for xVelocity
- Used in Power BI, SSAS, and Power Pivot
- Adds meaning to data through expressions and formulas
- Adds calcuations to the model: calculated columns, measures, calculated tables, row filters
- Feels like Excel formulas, but **model-aware**, meaning itreferences table/column names
  instead of cell coordinates

### Connection to Previous Weeks
| Power BI Tool | Analogy from Earlier Weeks |
|---|---|
| Power Query | Like Pandas — reshape, clean, rename |
| Power BI Desktop | Like Jupyter — your local working environment |
| Power BI Service | Like GitHub Pages — publish for others to see |
| DAX | Like SQL aggregations — SUM, COUNT, FILTER, but inside the model |

<h2 style="color:steelblue">Section 1.B: Power BI Tour</h2>

### 1.B.1 & B.2 Power BI Versioning Strategy

- Microsoft constantly updates Power BI
- To share files the best way is to always be on the latest version
- How to check version: **Help ribbon → About**
- Always update

### 1.B.3 Home Screen

Appears when Power BI Desktop first loads. Used to:
- Start a **blank report** or connect directly to a data source
- **Open recent** .pbix projects
- Access **Recommended** Microsoft Learn training content

### 1.B.4 The Ribbon

- Tab bar across the top of the interface
- **Context-sensitive** — available tabs and options change based on:
  - Which **View** is currently active
  - What **object** is selected on the canvas

### 1.B.5 - B.18 The Ribbon Tabs & The Four Views

The ribbon is the tab bar at the top of Power BI Desktop. Think of it like a toolbox — 
different tools appear depending on where you are and what you have selected. 
Here is what each tab does and when it shows up.

---

### 🗂️ Ribbon Tabs
The ribbon is the **context-sensitive toolbox** at the top — what you see changes based
on your active view and what's selected.

| Tab | Available In | What It Does |
|---|---|---|
| **File** | Always | Manage project files, access Options & Settings |
| **Home** | Always | Core actions — get data, refresh, publish |
| **Insert** | Report View only | Add visuals, text boxes, buttons to a report page |
| **Modeling** | Report View only | Shortcuts to create measures, manage relationships |
| **View** | Report View only | Control editor appearance — themes, gridlines, panels |
| **Format** | Report View (contextual) | Format a selected object on the canvas |
| **Data/Drill** | Report View (contextual) | Drill into data of a selected visual |
| **Table Tools** | Report + Table View (contextual) | Actions on a selected table |
| **Help** | Report, Table, Model | Same options in every view — use to check your version |

> **Contextual** means the tab only appears when something specific is selected.

---

### The Four Views

| View | Icon | Primary Use |
|---|---|---|
| **Report View** | 📊 | Design and view your report pages |
| **Table View** | 🔲 | Browse raw data; modify columns |
| **Model View** | 🔗 | See and configure relationships between tables |
| **DAX Query View** | DAX | Query your model with DAX, like SQL for your data |

**Report View** has 4 panels to know:
- **Pages Area** — add/switch report pages, like tabs in Excel
- **Filters Pane** — control what data is shown per page or per visual
- **Visualizations Pane** — pick a chart type and map data to it
- **Data/Fields Pane** — browse all tables and fields to drag into visuals

**Model View** tip — you can add extra pages to focus on a subset of tables,
useful when your model gets large.

**DAX Query View** tip — queries are saved to the model but won't show up
after publishing to the Power BI Service.
