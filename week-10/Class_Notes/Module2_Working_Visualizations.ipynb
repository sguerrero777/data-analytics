<h2 style="color:steelblue">Section 3.A: Customizing Visualizations and Report Interactions</h2>

### 3.A.24 Formatting Options

Formatting can be applied to **each specific visualization**. Options include titles and text, colors, gridlines, borders, and other visual elements. When a visual is selected, these live under the **Formatting tab** of the Visualizations pane (the paint-roller icon).

Options **vary by visualization type**, organized into two tabs:

| Tab | Scope |
|---|---|
| **Visual** | Options specific to the chosen visualization type (a bar chart shows axis, legend, data-label options) |
| **General** | Universal settings — title, background, border, size/position |

⚠️ **Formatting is non-destructive** — changes apply only to the visual; the source data is never altered.

> 💡 Explore all tabs when working with a new visual type, since each chart exposes different options. This is pure presentation — nothing here touches the model or the data, unlike M and DAX.

### 3.A.25 Report Themes

A **theme** applies a color scheme across the **entire report** at once.

- Accessed from the **View tab** of the ribbon.
- All visuals conform to the theme **by default**, unless a visual has its own custom formatting (custom formatting wins).
- **Custom themes** can be built from branding guidelines, then **exported and imported** to reuse across reports and workspaces.

> 💡 Theme = report-wide styling in one click; per-visual formatting (3.A.24) = one visual at a time. Think global stylesheet vs inline styles. ⭐ Standout tip: a custom theme is a small JSON file — building one to match a company's brand colors is a portfolio-worthy detail that shows you think about consistency, not just individual charts.

<h2 style="color:steelblue">Section 3.B: Report Interactions</h2>

### 3.B.1 Filter Interactions

By default, **filter interactions apply across a report page** as the user interacts. Clicking an attribute of one visual filters the whole page to that attribute — this is **cross-filtering**, on by default.

The **Filters pane** adds three scopes of filtering control:

| Filter Scope | Applies To |
|---|---|
| **Visual-level** | Only the selected visual |
| **Page-level** | Every visual on the current report page |
| **Report-level** | Persists across **all** pages in the report |

Filters can be set to **Basic (list)**, **Advanced (conditions)**, or **Top N** mode.

⭐ Tip: **lock or hide** a filter in the Filters pane so end-users can't change it.

> 💡 Cross-filtering lets users explore data contextually **without writing DAX** — one of Power BI's most powerful UX features. It's like clicking a value and having every other chart auto-`WHERE` itself to match.

### Cross-Filtering — Demo

Using the **RetailSales_Module3** table (the lab dataset: Category, CustomerSegment, Discount, OrderDate, OrderID, Product, Profit, Quantity, Region, Salesperson, ShipMode, SubCategory, TotalSales, UnitPrice).

A "Sum of Profit by Region" line chart (Region on X-axis, Sum of Profit on Y-axis) declines West → North → East → South. Clicking a region cross-filters every other visual on the page to that region. The Filters pane shows the three scopes: **Filters on this visual**, **Filters on this page**, **Filters on all pages**.

> 💡 This is the visual-level / page-level / report-level split from 3.B.1, seen live in the Filters pane.

### 3.B.2 Move Down vs. Expand Hierarchy

A hierarchy can be **drilled into** or **expanded**, breaking data into more granular levels. Example hierarchies: Year → Quarter → Month, or Category → Subcategory → Product.

| **Go to Next Level** (double-arrow ⬇⬇) | **Expand** (forked-arrow) |
|---|---|
| Moves the **entire** visual down one level | **Keeps** parent-level categories, adds the next level beneath |
| All top-level categories **replaced** by the next level | Parent stays visible; each expands to show children |
| Example: Year → Quarter (all years collapse, all quarters shown) | Example: each Year expands to show its Quarters |
| Use when focusing **entirely** on the next granularity | Use when **comparing** parent and child levels together |
| | More detailed, layered view |

Demo: "Sales by Category" → **Go to next level** → "Sales by Product" (categories replaced). vs → **Expand** → "Sales by Category and Product" (products nested under each category).

> 💡 Move Down = swap the level. Expand = add the level underneath. The forked-arrow keeps both layers; the double-arrow trades up for down.

### 3.B.3 Drill Down, Navigate Up

**Drill Down** (single arrow — off/on toggle) lets you click a **single element** of a chart and filter down to the next hierarchy level **for that element only** (not the whole visual).

To go back **up** a level, use the **up arrow** — and it works no matter how you moved down (drill, move-down, or expand).

> 💡 Key difference from 3.B.2: Move Down / Expand change the *whole* visual's level; **Drill Down zooms into one bar/slice you click**. Like double-clicking a single folder to see inside it, vs changing the view of every folder at once. The up arrow is your universal "back."

<h2 style="color:steelblue">Section 3.C: Data-based Formatting</h2>

### 3.C.1 Customizing Visuals with Data

Static formatting isn't enough — **let the data drive how the report looks**. Three data-driven techniques:

| Technique | What It Does |
|---|---|
| **Conditional Formatting** | Applies colors, icons, or data bars based on value thresholds |
| **Grouping & Binning** | Organizes categories or ranges of values for cleaner views |
| **Clustering** | Uses **machine learning** to auto-detect groups in scatter plots |

Why use it: draws the eye to important values automatically, makes patterns visible without heavy analysis, and **updates automatically as data refreshes** — no redesign needed.

> 💡 Combining all three can turn a plain table into an insight-rich visual in minutes. This is the report-layer echo of everything before it: instead of *you* computing what matters (DAX) or cleaning it (M), the *formatting* surfaces it visually.

### 3.C.2 Conditional Formatting

Lets you **visually highlight** important data or trends in a report or dashboard. Formatting is based on **data values** — color-coding columns, or adding **data bars** or **icons** to a table.

> 💡 This is where the DAX from Module 2 pays off: conditional formatting can be driven **by a measure** (e.g. color a cell red when a "% to target" measure drops below a threshold). ⭐ Standout tip: "format by a measure" (using `SWITCH` or a rules-based DAX measure to return color codes) is an advanced move most beginners skip — it makes formatting fully dynamic and reactive to filter context.

### 3.C.3 Grouping

Grouping lets users **combine data based on common attributes or characteristics**. It helps **simplify complex datasets** by reducing the number of categories or values displayed.

Demo: on a "Sum of Profit by Month Name" chart, select data points → right-click → **Group > Group** (other menu items: Show as a table, Analyze, Summarize, Copy).

> 💡 Like a manual `GROUP BY` you build by selecting items — e.g. bucket several small product categories into one "Other" group to declutter a chart.

### 3.C.4 Binning

Binning lets you set a **bin size** for **numeric or time** fields — turning continuous values into ranges.

- Number and size of bins are customizable, along with the **ranges and labels** for each bin.
- Created using **"New group"** and managed in the **Groups dialog**.
- ⚠️ You can bin **calculated columns but NOT measures**.

> 💡 Binning is `pd.cut()` — chopping a continuous column (ages, prices) into buckets like "0–18, 19–35, 36+". The "can't bin a measure" rule ties back to Module 2: a measure has no fixed per-row value to bucket, it's computed live per filter context.

### 3.C.5 Clustering

Available for **scatter plot** visualizations, clustering uses **machine learning** to group data points with similar characteristics — surfacing patterns that may not be obvious. Found under the **`...` menu** at the corner of a selected scatter plot ("Automatically find clusters").

Demo: a "Sales Quantity vs Product Unit Price" scatter → **Automatically find clusters** → points auto-colored into Cluster1, Cluster2, Cluster3.

> 💡 This is literally **k-means** from scikit-learn, but Power BI runs it from a right-click — no `from sklearn.cluster import KMeans` needed. ⭐ Standout tip: knowing it's k-means under the hood means you can explain *why* it only works on scatter plots (it needs two numeric axes as features) — that's the kind of "I understand the mechanism" insight that stands out.

<h2 style="color:steelblue">Section 3.D: Adding Interactive Features</h2>

### 3.D.1 Customizing Filter Interactions

Default cross-filter behavior can create **unintended results** on complex pages — e.g. a region slicer accidentally filtering a KPI card that should stay static. Multiple filters can also compete or get overly granular in combination.

**Edit Interactions** mode lets you control how a selected visual filters every other visual. Access: select a visual → **Format tab → Edit Interactions**. Each other visual then shows a control icon in its top-right corner:

| Icon | Behavior |
|---|---|
| **Filter** (funnel) | Selected visual filters this one (default) |
| **Highlight** (bar chart) | Highlights matching data **without** filtering others out |
| **None** (circle with slash) | This visual is **completely unaffected** |

Customizations are **saved with the report** and affect all viewers.

⭐ Tip: set **None** on summary KPI cards so they always show totals regardless of page-level filter selections.

> 💡 This is the fix for "why did my total card change when I clicked a bar?" — Edit Interactions lets you protect the visuals that should stay put.

### 3.D.2 Customizing Tooltips

Tooltips can be customized in **how and what** they display — from simple formatting changes to adding a **custom tooltip page** to the report.

- Custom tooltips provide **more detailed info** about specific data points.
- They enhance interactivity and can support **accessibility** (audio or descriptive text).

Demo: hovering a month on a line chart shows a default tooltip (Direct/Indirect/Logistics/Other totals), upgraded to a richer tooltip plus a **Sub Category breakdown table** on hover.

> 💡 A custom tooltip page is a mini-report that pops up on hover — like a detail panel that appears without the user clicking away. ⭐ Standout tip: building a tooltip *page* (not just formatting the default) lets you show a whole mini-chart on hover, a polished touch that signals report-design maturity.

### 3.D.3 Drill Mode

Turning on **Drill Mode** for a visual makes clicking a data point trigger a **drill-down** interaction.

- Toggle it with the **down arrow** in the corner of the visual.
- Once a drill-down filter is applied, **drill up** with the same up arrow used for hierarchy na|vigation.

> 💡 Drill Mode is the on/off switch that decides whether a click *drills* (zooms into that data point's next level) or just *cross-filters* the page. It connects 3.B.3's drill-down to a clickable mode.

### 3.D.4 Drill-through Pages

Drill-through pages let users **explore a specific data point in depth** without cluttering the main page.

- Enabled by adding a field (used elsewhere in the report) to the **drill-through well** of *another* report page.
- Filters from the **starting page pass through** to the drill-through page unless you specify otherwise.

> 💡 Right-click a data point → "Drill through" → jump to a dedicated detail page already filtered to that point. Like clicking a row in a summary table to open its full record. ⭐ Standout tip: drill-through + a clean detail page is exactly the "summary-to-detail" pattern hiring managers love to see in a portfolio report.

### 3.D.5 Bookmarks

Bookmarks **capture and navigate to specific states** of a report, including applied filters.

- Create dynamic displays of different scenarios or perspectives.
- Provide convenient shortcuts to key insights.
- Toggle the Bookmarks panel from the **View tab**; add and manage bookmarks there.

> 💡 A bookmark is a saved snapshot of the report's current state (filters, selections, even which visuals are visible). ⭐ Standout tip: bookmarks + buttons let you build a **navigation experience** (toggle between chart and table views, story-style "next" buttons) — that's how you turn a report into something that feels like an app.

### 3.D.6 Q&A Visualization — ⚠️ Deprecated

The **Q&A visualization** let users ask questions about report data in **plain language** ("show sales by region last year"), understanding synonyms and word variations. Synonyms could be added for specialized vocabulary. It could generate tables, charts, maps, and timelines.

⚠️ **It's being discontinued by Microsoft** and will no longer be supported, though it still appears in some sample reports.

> 💡 Worth knowing it exists (you'll see it in samples) but **don't build new reports around it**. Knowing what's deprecated is itself a standout signal — it shows you're current, not learning from stale tutorials.

### 3.D.7 Custom Visualizations

Beyond the built-in visuals, more are available via **AppSource**, from Microsoft and other providers:

- **Chiclet slicer** (uses custom images)
- **Gantt, Sankey, sunburst, tornado**, and other chart types
- **Enhanced KPIs and gauges**
- **Animated** charts and slicers for engagement

> 💡 When the built-in gallery can't show what you need (e.g. a Sankey flow diagram), AppSource is where you get it. ⭐ Standout tip: knowing a few named custom visuals (Sankey for flows, sunburst for hierarchies) lets you pick the *right* chart for a data story instead of forcing everything into a bar chart.

### 🔁 Confusion Corner

**"Move Down vs Expand vs Drill Down?"** ⭐ Move Down = whole visual drops a level (parents replaced). Expand = whole visual keeps parents AND adds children beneath. Drill Down = click ONE element and zoom into just that one's next level. Up arrow reverses any of them.

**"Theme vs visual formatting?"** Theme = report-wide colors (View tab, one click). Visual formatting = one visual at a time (paint-roller, Visual/General tabs).

**"Filter vs Highlight vs None (Edit Interactions)?"** Filter = removes non-matching data (default). Highlight = dims non-matching but keeps it visible. None = the visual ignores the selection entirely (use on KPI cards).

**"Grouping vs Binning?"** Grouping combines **categories** (bucket products into "Other"). Binning chops **numeric/time ranges** into buckets (ages into 0–18, 19–35...). And you can bin a calculated column but not a measure.

**"Drill Mode vs Drill Down vs Drill-through?"** Drill Mode = the on/off toggle that makes clicks drill. Drill Down = going one level deeper on a clicked element. Drill-through = jumping to a separate detail *page* for that data point.

**"Is conditional formatting just colors?"** No — colors, icons, AND data bars, all based on value thresholds, and it can be driven by a DAX measure for fully dynamic formatting.

### ⭐ When Would I Use This? (Module 3)

**Formatting & themes:**
- Match a report to company branding with a custom theme (export/import to reuse).
- Clean up one chart's axis/legend/labels via the Visual tab.

**Filters & interactions:**
- Cross-filter so clicking a region updates the whole page.
- Set a KPI card to **None** so it always shows the grand total.
- Lock a filter so end-users can't change it.

**Hierarchy navigation:**
- Move Down to swap Year → Quarter; Expand to show Quarters under each Year; Drill Down to zoom into one region only.

**Data-based formatting:**
- Conditional formatting (driven by a measure) to flag underperforming cells red.
- Group small categories into "Other"; bin ages or prices into ranges.
- Clustering to auto-find customer segments in a scatter plot.

**Interactive features:**
- Custom tooltip page to show a mini-chart on hover.
- Drill-through page for summary → detail.
- Bookmarks + buttons to build app-like navigation or toggle chart/table views.
- AppSource visuals (Sankey, sunburst) when built-ins can't tell the story.

### Key Takeaways 🎯

Module 3 is the **report layer** — after M cleaned the data and DAX analyzed it, this is how the numbers actually reach a human. Less code, more design and interactivity.

**Making visuals look right:**
Each visual formats independently (paint-roller → **Visual** tab for chart-specific options, **General** tab for universal ones), and it's all **non-destructive** — never touches the data. Want report-wide consistency in one click? Apply a **theme** from the View tab (custom themes export/import for reuse).

**Letting users explore:**
**Cross-filtering** is on by default — click anything, the page reacts, no DAX needed. Filters come in three scopes (visual / page / report) and three modes (Basic / Advanced / Top N). For hierarchies: **Move Down** swaps the level, **Expand** adds a level underneath, **Drill Down** zooms into one clicked element, and the **up arrow** always takes you back.

**Letting the data drive the look:**
**Conditional formatting** (colors/icons/data bars by threshold — bonus points for driving it with a DAX measure), **grouping** (combine categories), **binning** (`pd.cut` for numeric/time ranges — columns only, not measures), and **clustering** (k-means from a right-click, scatter plots only).

**Advanced interactivity:**
**Edit Interactions** decides whether a visual filters, highlights, or ignores a selection (set KPI cards to **None**). **Custom tooltips** can be whole mini-report pages. **Drill-through** jumps to a filtered detail page. **Bookmarks** save report states for app-like navigation. **Q&A** exists but is **deprecated** — don't build on it. And **AppSource** has custom visuals (Sankey, sunburst, Gantt) when the built-ins fall short.
