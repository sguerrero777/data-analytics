## Array Creation

| Function | Syntax | Example | Result |
|---|---|---|---|
| `array()` | `np.array(list)` | `np.array([1,2,3])` | `[1 2 3]` |
| `zeros()` | `np.zeros(shape)` | `np.zeros((2,3))` | 2×3 grid of 0s |
| `ones()` | `np.ones(shape)` | `np.ones((2,2))` | 2×2 grid of 1s |
| `full()` | `np.full(shape, value)` | `np.full((2,2), 7)` | `[7 7; 7 7]` |
| `arange()` | `np.arange(start, stop, step)` | `np.arange(0,10,2)` | `[0 2 4 6 8]` |
| `linspace()` | `np.linspace(start, stop, n)` | `np.linspace(0,1,5)` | `[0 0.25 0.5 0.75 1]` |
| `empty()` | `np.empty(shape)` | — | Uninitialized array |
| `eye()` | `np.eye(n)` | — | Identity matrix |
| `random.rand()` | `np.random.rand(shape)` | — | Random values 0–1 |
| `random.randint()` | `np.random.randint(low, high)` | — | Random integers in range |

```python
import numpy as np

print("=== NumPy Array Creation Functions ===\n")

# 1. array() - create from Python list
a = np.array([1, 2, 3, 4, 5])
print("np.array([1,2,3,4,5])")
print(a, "\n")

# 2. zeros() - array of zeros
b = np.zeros((2, 3))
print("np.zeros((2,3))")
print(b, "\n")

# 3. ones() - array of ones
c = np.ones((2, 2))
print("np.ones((2,2))")
print(c, "\n")

# 4. full() - array filled with a constant value
d = np.full((3, 2), 7)
print("np.full((3,2), 7)")
print(d, "\n")
```

## Array Dimensions — 1D, 2D, 3D

```python
import numpy as np

# 1D Array
arr_1 = np.array([1, 2, 3])
print(arr_1)

# 2D Array
arr_2 = np.array([
    [1, 2, 3],
    [1, 2, 3],
    [1, 2, 3]
])
print(arr_2)

# 3D Array
arr_3 = np.array([
    [[1, 2, 3],
     [1, 2, 3],
     [1, 2, 3]],
    [[1, 2, 3],
     [1, 2, 3],
     [1, 2, 3]],
    [[1, 2, 3],
     [1, 2, 3],
     [1, 2, 3]]
])
print(arr_3)
```

## Array Attributes

| Attribute | Meaning | Example |
|---|---|---|
| `ndim` | Number of dimensions | 1D, 2D, 3D |
| `shape` | Size of each dimension (rows, columns) | `(3, 4)` |
| `size` | Total number of elements | `12` |
| `dtype` | Data type of elements | `int64`, `float64` |
| `itemsize` | Memory size (bytes per element) | `8` |
| `nbytes` | Total memory used | `96` |
| `T` | Transpose (rows ↔ columns) | flipped matrix |
| `real` | Real part (for complex numbers) | real values |
| `imag` | Imaginary part | imaginary values |
| `data` | Buffer storing raw data | memory reference |

```python
import numpy as np

# Basic operations on a 1D array
a = np.array([10, 20, 30, 40])
print(a * 2)      # [20, 40, 60, 80]
print(a.mean())   # 25.0
print(a.dtype)    # int64

# Attributes on a 2D array
a = np.array([[10, 20, 30],
              [40, 50, 60]])

# .shape - dimensions (rows, columns)
print("\nShape of array:", a.shape)
# shows rows and columns

# .ndim - number of dimensions
print("Number of dimensions (ndim):", a.ndim)
# 2

# .dtype - data type of elements
print("Data type (dtype):", a.dtype)

# .size - number of elements
print("Size of array:", a.size)
```

## Looping Through Arrays

```python
import numpy as np

arr2d = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

# Loops through rows then elements
for row in arr2d:
    print("Row:", row)

# first loop gets rows
# second loop gets elements inside each row
for row in arr2d:
    for value in row:
        print(value)
```

```python
import numpy as np

# 3D full traversal — loop tru the elements in 3D
arr3d = np.array([
    [
        [1, 2],
        [3, 4]
    ],
    [
        [5, 6],
        [7, 8]
    ]
])

# Full traversal
for layer in arr3d:
    for row in layer:
        for value in row:
            print(value)
# This code is doing a full traversal of a 3D NumPy array,
# breaking it down level by level until it reaches individual numbers
```

## Joining / Concatenating Arrays

| Operation | What happens | Result |
|---|---|---|
| `concatenate` | Joins arrays | `[1 2 3 4 5 6]` |
| `+` operator | Element-wise addition | `[5 7 9]` |

***if you use concate operator, values of first array get appended to the other***

***+ operator adds everything in the array with their respective indexes***



### Stacking Helper Functions

| Function | Purpose | Axis | Syntax |
|---|---|---|---|
| `np.concatenate()` | Join along existing axis | flexible (0,1,2...) | `np.concatenate((a,b), axis=...)` |
| `np.stack()` | Join into new dimension | creates new axis | `np.stack((a,b), axis=0)` |
| `np.vstack()` | Vertical / row-wise | axis=0 | `np.vstack((a,b))` |
| `np.hstack()` | Horizontal / column-wise | axis=1 | `np.hstack((a,b))` |
| `np.dstack()` | Depth / 3rd dimension | axis=2 | `np.dstack((a,b))` |
| `np.column_stack()` | Stack as columns | 1D → columns | `np.column_stack((a,b))` |
| `np.row_stack()` | Stack as rows | like vstack | `np.row_stack((a,b))` |

```python
import numpy as np

# 1D concatenate
arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])
arr3 = np.concatenate((arr1, arr2))
print(arr3)
```

```python
import numpy as np

arrA = np.array([[1, 2], [3, 4]])
arrB = np.array([[5, 6], [7, 8]])

# axis=0 = join vertically (row-wise) → shape becomes (4x2)
result1 = np.concatenate((arrA, arrB), axis=0)
print(result1)

# axis=1 = join horizontally (column-wise) → shape becomes (2x4)
result2 = np.concatenate((arrA, arrB), axis=1)
print(result2)
```
