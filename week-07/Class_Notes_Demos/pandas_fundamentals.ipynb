## Pandas Data Structures

Pandas provides two primary data structures: **Series** and **DataFrame**

### Series
A Pandas Series is a one-dimensional labeled array. It can be created from many data types, most commonly a Python list or Python dictionary.

> **A Series is like a single column of a spreadsheet — each value has a label/index**

### Series Functions Reference

| Function/Method | Description | Example Syntax | Example Output |
|---|---|---|---|
| `pd.Series()` | Creates a Series object | `s = pd.Series([10,20,30])` | `0 10 1 20 2 30` |
| `.head()` | Displays first rows | `s.head(2)` | First 2 values |
| `.tail()` | Displays last rows | `s.tail(2)` | Last 2 values |
| `.dtype` | Shows data type | `s.dtype` | `int64` |
| `.shape` | Returns dimensions | `s.shape` | `(3,)` |
| `.size` | Returns number of elements | `s.size` | `3` |
| `.index` | Displays index labels | `s.index` | `RangeIndex(...)` |
| `.values` | Returns NumPy array values | `s.values` | `[10 20 30]` |
| `.sum()` | Calculates total | `s.sum()` | `60` |
| `.mean()` | Calculates average | `s.mean()` | `20.0` |
| `.median()` | Finds middle value | `s.median()` | `20.0` |
| `.mode()` | Finds most frequent value | `s.mode()` | Most common value |
| `.min()` | Smallest value | `s.min()` | `10` |
| `.iloc[]` | Access by integer position | `s.iloc[0]` | First value |
| `.loc[]` | Access by label | `s.loc[1]` | Value at label 1 |
| `.append()` | Combines Series (deprecated) | `pd.concat([s1,s2])` | Combined Series |
| `.copy()` | Creates copy | `s.copy()` | Duplicate Series |

````python
# A Pandas Series is a one-dimensional labeled array. It can be created from many
# data types, most commonly a Python list or Python dictionary
import pandas as pd

# Creating a Pandas Series from a Python list
# A SERIES IS LIKE A SINGLE COLUMN OF A SPREADSHEET, EACH VALUE HAS A LABEL/INDEX
data = [10, 20, 30, 40]
# data is just a regular Python list — pd.Series() will convert it into a Series

s = pd.Series(data) # pd.Series() is the function that creates a Series object
print(s)
# the numbers on the left (0, 1, 2, 3) are the index — automatically assigned
# the numbers on the right (10, 20, 30, 40) are the values
````

````python
# Creating a Pandas Series from a Python Dictionary
# when using a dictionary, the keys become the index labels instead of numbers
data = {
    "Excel": 85,
    "SQL": 90,
    "Python": 78
}
# keys (Excel, SQL, Python) become the index
# values (85, 90, 78) become the data

# Creating a series called s
s = pd.Series(data)
print(s)

# when you create a series, it gives index values which you can reference back to
# for example: s["Excel"] would return 85
````

## DataFrame

| Concept | Meaning |
|---|---|
| Rows | Horizontal records |
| Columns | Vertical categories |
| Index | Row numbers (0,1,2...) |
| `pd.DataFrame()` | Creates a DataFrame |

### Ways to Create a DataFrame

| Method | Structure | Best Used When |
|---|---|---|
| List of Lists | Rows of data | Data already arranged row-by-row |
| Dictionary | Columns of data | Data organized by columns |

### Working with DataFrame

- **Inspect data:** `head()`, `info()`, `describe()`
- **Select data:** `loc[]`, `iloc[]`
- **Clean data:** `fillna()`, `dropna()`
- **Transform data:** `groupby()`, `sort_values()`, `astype()`
- **Combine data:** `merge()`, `concat()`

### Full DataFrame Function Reference

| Function | Syntax | Purpose | Example |
|---|---|---|---|
| `read_csv()` | `pd.read_csv("file.csv")` | Load data from a CSV file | `df = pd.read_csv("data.csv")` |
| `head()` | `df.head(n)` | Show first n rows | `df.head(5)` ||
| `tail()` | `df.tail(n)` | Show last n rows | `df.tail(3)` |

| `shape` | `df.shape` | Returns (rows, columns) | `df.shape` |

````python
# A DataFrame is a 2-dimensional table in Pandas made up of rows and columns.
# Each inner list becomes a row.
# Think of it like a spreadsheet — rows go across, columns go down

import pandas as pd

# Creating a DataFrame from a Python List (List of Lists)
# each inner list = one row of data
data = [
    ["Alice", 20],  # row 0
    ["Bob", 22],    # row 1
    ["Charlie", 21] # row 2
]

df = pd.DataFrame(data, columns=["Name", "Age"])
# pd.DataFrame() converts the list into a table
# columns=["Name", "Age"] assigns header names to each column
# without columns=, headers would default to 0 and 1

print(df)
# the numbers on the far left (0, 1, 2) are the index — automatically assigned
````

````python
import pandas as pd

# Creating a DataFrame from a Dictionary
# a dictionary is another common way to create a DataFrame
# the structure is different — you organize by column instead of by row
data = {
    "Name": ["Alice", "Bob", "Charlie"],  # Name column
    "Age": [20, 22, 21]                   # Age column
}
# Keys become column names
# Values become column data

df = pd.DataFrame(data)
# no need for columns= because the dictionary keys are already the column names
# Pandas automatically uses the keys as headers

print(df)
````

````python
# Insert a column using .insert()
# .insert() lets you add a new column at a specific position

import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie"],
    "SQL": [85, 78, 92]
}

df = pd.DataFrame(data)

print(df)
# currently has 2 columns: Name (position 0) and SQL (position 1)

print()
# insert a column
df.insert(1, "Java", [80, 85, 90])
# .insert(position, column name, values)
# position 1 means it goes between Name and SQL
# inserts a new "Java" column at position 1 (between Name and SQL) or index 1
# this modifies the existing df directly — no need to reassign

print(df)
# now has 3 columns: Name, Java, SQL — Java was inserted at position 1
````

## Indexing a DataFrame

Indexing in a DataFrame means **selecting rows, columns, or individual values** from a table.

Pandas provides two main indexing methods:
- `loc[]` → label-based indexing
- `iloc[]` → position-based indexing

````python
# Indexing a DataFrame
# indexing means selecting specific rows, columns, or values from a DataFrame

import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie", "David"],
    "Age": [20, 22, 21, 23],
    "Java": [85, 78, 92, 80]
}

df = pd.DataFrame(data)

print(df)
# the numbers on the left (0, 1, 2, 3) are the index labels

# Label-Based Indexing (loc[])
# loc[] uses the actual index label to find a row
# loc[] uses row labels and column names
# Select a Row
print()

print(df.loc[1])
# df.loc[1] selects the row where the index label is 1 (Bob's data)
# returns all columns for that row as a Series
# Name: Bob, Age: 22, Java: 78
````

````python
# Position-Based Indexing (iloc[])
# iloc[] uses row and column positions (0-based indexing)
# think of it like coordinates — [row number, column number]
# counting always starts at 0

import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie", "David"],
    "Age": [20, 22, 21, 23],
    "Java": [85, 78, 92, 80]
}

df = pd.DataFrame(data)

print(df)

# Select a Row
print(df.iloc[1])
# iloc[1] selects the row at position 1 (Bob) — 0=Alice, 1=Bob, 2=Charlie, 3=David

print()
# Select a Specific Value
print(df.iloc[2, 0])
# iloc[2, 0] means row 2, column 0 → Charlie (row 2) Name (column 0)

print()
# Select Multiple Rows and Columns
print(df.iloc[1:3, 0:2])
# iloc[1:3, 0:2] means rows 1 and 2, columns 0 and 1
# 1:3 goes from row 1 up to but NOT including row 3
# 0:2 goes from column 0 up to but NOT including column 2

# Select One Column
print()
print(df["Name"])
# df["Name"] selects the entire Name column as a Series

print()
# Select Multiple Columns
print(df[["Name", "Java"]])
# df[["Name", "Java"]] selects multiple columns — notice the double brackets
# double brackets returns a DataFrame, single bracket returns a Series
````

## Statistical Functions

Pandas statistical functions are used for aggregation — they take a whole column of values and reduce them down to a single number.

| Function | Description |
|---|---|
| `.sum()` | Adds all values |
| `.count()` | Counts how many values |
| `.mean()` | Calculates the average |
| `.median()` | Finds the middle value |
| `.min()` | Finds the smallest value |
| `.max()` | Finds the largest value |
| `.agg()` | Run multiple aggregations at once |

````python
# Pandas Statistical Functions
# these functions are called aggregation functions
# they take a whole column of values and reduce them down to a single number
import pandas as pd

data = {
     "Student":["Amy", "Bob", "Cara", "Dan", "Eva"],
     "Score":[85, 92, 78, 95, 100]
 }
# Creates a dataframe referenced by df variable
df = pd.DataFrame(data)
print(df)
print()

print("SUM:", df["Score"].sum())
# .sum() adds all values in the Score column together
# df["Score"] selects just the Score column as a Series

print("COUNT:", df["Score"].count())
# .count() counts how many non-null values are in the column
# useful for checking if any values are missing

print("MEAN:", df["Score"].mean())
# .mean() calculates the average score

print("MEDIAN:", df["Score"].median())
# .median() finds the middle value when sorted
# less affected by outliers than mean

print("MIN:", df["Score"].min())
# .min() returns the lowest score

print("MAX:", df["Score"].max())
# .max() returns the highest score

print("\nAggregate Results:")
print(df["Score"].agg(["sum", "mean", "median", "min", "max"]))
# .agg() lets you run multiple aggregations at once
# pass a list of function names as strings
# returns all results together in one output
````

## Sorting and Ranking

- `sort_index()` → sorts the DataFrame by its index
- `sort_values()` → sorts by column values
- `rank()` → assigns a rank to each value; `ascending=False` gives rank 1 to the highest value

````python
# sort_index() and rank()
# display result in descending

import pandas as pd

# Create DataFrame
# notice index=[3, 1, 2, 0] — we are manually setting a custom index
# instead of the default 0, 1, 2, 3 the index will be 3, 1, 2, 0
df = pd.DataFrame({
    "Student": ["Amy", "Bob", "Cara", "Dan"],
    "Score": [88, 95, 79, 91]
}, index=[3, 1, 2, 0])

print("ORIGINAL DATAFRAME")
print(df)
# notice rows are displayed in the order the index was given (3, 1, 2, 0)
print()

# Sort index from highest to lowest
print("\nSORTED INDEX (Highest to Lowest)")
sorted_df = df.sort_index(ascending=False)
# sort_index() reorders the rows based on their index value
# ascending=False sorts the index from highest to lowest (3, 2, 1, 0)
# ascending=True (default) would sort lowest to highest (0, 1, 2, 3)
print(sorted_df)

# Rank scores from highest to lowest
df["Rank"] = df["Score"].rank(ascending=False)
# .rank() assigns a rank number to each value
# ascending=False gives rank 1 to the highest score, rank 2 to second highest, etc.
# a new column called "Rank" is added to the existing df

# Sort by rank so highest score appears first
ranked_df = df.sort_values(by="Rank")
# sort_values(by="Rank") sorts the entire DataFrame by the Rank column
# lowest rank number (1) appears first

print("\nRANKINGS (Highest to Lowest)")
print(ranked_df)
````

## Merging DataFrames in Pandas

Merging combines two DataFrames together using a common column. It works like joining tables in a database.

| Merge Type | Description |
|---|---|
| `inner` | Only matching rows |
| `left` | All rows from left DataFrame |
| `right` | All rows from right DataFrame |
| `outer` | All rows from both DataFrames |

### pd.merge() Parameters

| Function | Purpose |
|---|---|
| `pd.merge()` | Combine DataFrames |
| `on=` | Common column |
| `how="inner"` | Matching rows only |
| `how="left"` | Keep all left rows |
| `how="right"` | Keep all right rows |
| `how="outer"` | Keep all rows |
| `left_on=` | Left column name |
| `right_on=` | Right column name |

> **Note:** When merging, if both DataFrames have a column with the same name (other than the key), Pandas will automatically rename them to `_x` and `_y` to avoid confusion. Also, `NaN` appears wherever there is no matching data in the other DataFrame.

````python
# merging dataframes
# merging combines two DataFrames using a common column
# works like joining tables in a database

import pandas as pd

# DataFrame 1: Students
# contains StudentID and Name
students = pd.DataFrame({
    "StudentID": [1, 2, 3, 4],
    "Name": ["Amy", "Bob", "Cara", "Dan"]
})

# DataFrame 2: Scores
# contains StudentID and Score
# notice StudentID 4 (Dan) is missing but StudentID 5 exists
# this means Dan has no score, and StudentID 5 has no name
scores = pd.DataFrame({
    "StudentID": [1, 2, 3, 5],
    "Name": [85, 92, 78, 88]
})

print("STUDENTS DATAFRAME")
print(students)
print("\nSCORES DATAFRAME")
print(scores)
print()

# 1. Inner Join
# only keeps rows where StudentID exists in BOTH DataFrames
# Dan (ID 4) and ID 5 are excluded because they don't match in both
inner_merge = pd.merge(students, scores, on="StudentID", how="inner")
print("\nINNER MERGE (only matching StudentID)")
print(inner_merge)
# Name_x = Name from students DataFrame
# Name_y = Name from scores DataFrame
# Pandas adds _x and _y when both DataFrames share a column name

# 2. Left Join
# keeps ALL rows from the left DataFrame (students)
# if a student has no matching score, NaN fills the missing spot
left_merge = pd.merge(students, scores, on="StudentID", how="left")
print("\nLEFT MERGE (all students kept)")
print(left_merge)
# Dan (ID 4) appears with NaN for score because he has no match in scores

# 3. Right Join
# keeps ALL rows from the right DataFrame (scores)
# if a score has no matching student, NaN fills the missing spot
right_merge = pd.merge(students, scores, on="StudentID", how="right")
print("\nRIGHT MERGE (all scores kept)")
print(right_merge)
# StudentID 5 appears with NaN for Name because it has no match in students

# 4. Outer Join
# keeps ALL rows from BOTH DataFrames
# NaN fills wherever there is no matching data on either side
outer_merge = pd.merge(students, scores, on="StudentID", how="outer")
print("\nOUTER MERGE (all data from both)")
print(outer_merge)
# Dan (ID 4) has NaN for score, StudentID 5 has NaN for Name
````
