## `fontdict=` — Font Customization

Pass a dictionary to `fontdict=` on `plt.title()`, `plt.xlabel()`, or `plt.ylabel()`:

```python
font1 = {'color': 'darkblue', 'size': 15}
font2 = {'family': 'cursive'}
# families: 'serif', 'sans-serif', 'monospace', 'fantasy', 'cursive'

plt.title("My Chart", fontdict=font1, loc='right')
# loc options: 'left', 'center' (default), 'right'
plt.xlabel("X Label", fontdict=font2)
plt.ylabel("Y Label", fontdict=font2)
```

## What is Matplotlib?

**Key Concept:** Matplotlib is a powerful open-source Python library for creating static, animated, and interactive visualizations. Originally developed by John D. Hunter, it is widely used in data science, machine learning, and scientific computing for producing high-quality plots. It supports a variety of chart types, from simple line graphs to complex 3D visualizations, and integrates seamlessly with NumPy, Pandas, and Jupyter Notebooks.

**Two Main Interfaces:**
- **Pyplot API** — A MATLAB-like interface for quick plotting (what we use most: `plt.plot()`, `plt.show()`, etc.)
- **Object-Oriented API** — Provides fine-grained control over figures and axes (more advanced)

**Source:** [Matplotlib — Visualization with Python](https://matplotlib.org/)

## Labwork Reference: Chart-Specific Parameters

### Bar Charts (`plt.bar()` / `plt.barh()`)
- `color=` — bar fill color
- `width=` — bar width for vertical bars
- `height=` — bar width for horizontal bars (`barh`)
- First arg = categories array, second = values array

### Scatter Charts (`plt.scatter()`)
- `c=` — single color string OR numeric array (for heat map)
- `cmap=` — color map name, e.g. `'winter'`, `'plasma'`, `'viridis'`
- `plt.colorbar()` — adds a color scale legend next to the plot
- Multiple `scatter()` calls on same figure = multiple overlapping plots

### Histograms (`plt.hist()`)
- `bins=` — number of intervals/bars (more bins = more detail)
- `edgecolor=` — border color around each bar
- `alpha=` — transparency (0–1), useful when overlapping

### Pie Charts (`plt.pie()`)
- `labels=` — category name for each slice
- `startangle=` — rotate starting position (e.g. `90` = start at top)
- `explode=` — array to pull out slices, e.g. `[0.1, 0, 0, 0]`
- `shadow=True` — adds a drop shadow
- `colors=` — custom color list
- `plt.legend(title="Header")` — adds a legend with optional title

### Line Styles
| Code | Style |
|------|-------|
| `'-'` | Solid line (default) |
| `'--'` | Dashed line |
| `'-.'` | Dash-dot line |
| `':'` | Dotted line |

### Single-Letter Color Shortcuts (for fmt string)
| Code | Color |
|------|-------|
| `'b'` | Blue |
| `'g'` | Green |
| `'r'` | Red |
| `'c'` | Cyan |
| `'m'` | Magenta |
| `'y'` | Yellow |
| `'k'` | Black |
| `'w'` | White |

### Colors WITHOUT Single-Letter Shortcuts (use full name or hex)
- `color='pink'` or `color='#FFC0CB'`
- `color='orange'` or `color='#FFA500'`
- `color='purple'` or `color='#800080'`
- `color='brown'` or `color='#A52A2A'`
- These must be passed as a separate `color=` parameter — they cannot be used inside the fmt string

### fmt String Format
`'[color][marker][linestyle]'` — e.g. `'rs--'` = red, square marker, dashed line
- Fine-tune with keyword arguments: `ms` (marker size), `mfc` (marker face color)
- Use `color=` separately when the color has no single-letter shortcut

## Subplots

`plt.subplots()` returns two objects:
- **`fig`** — the entire figure container
- **`axes`** — array of individual plot objects

```python
fig, axes = plt.subplots(rows, cols, figsize=(width, height))

# Access each subplot by index:
axes[0].plot(x1, y1, color='blue')   # left / first plot
axes[1].bar(x2, y2, color='red')     # right / second plot

plt.show()
```

>  For a 2×2 grid use `axes[row][col]` — e.g. `axes[0][1]` = top right

## Saving a Figure

```python
plt.savefig('my_chart.png')           # save to PNG
plt.savefig('my_chart.png', dpi=150)  # higher resolution (150–300 for reports)
plt.savefig('my_chart.pdf')           # other formats: .jpg, .pdf, .svg
```

>  **Always call `savefig()` BEFORE `plt.show()`** — after `show()` the figure clears and you'll save a blank image!

## PyPlot Tutorial

````python
import matplotlib.pyplot as plt

# Displays the visualization directly inside the notebook
# (without this, the chart might open in a separate window)
%matplotlib inline

# Creating Python lists for data
# x-axis = categories (months), y-axis = values (sales numbers)
month = ['Jan', 'Feb', 'March', 'April', 'May', 'June']
sales = [100, 200, 400, 500, 150, 250]

# plt.bar() creates a vertical bar chart
# first argument = x-axis (categories), second = y-axis (values)
plt.bar(month, sales)       # Takes data from x, y and draws the bars
plt.title('Sales Over Month') # Sets the chart title
plt.xlabel('month')           # Labels the horizontal axis
plt.ylabel('sales')           # Labels the vertical axis
plt.show()                    # Displays the chart
````

### Line Graphs

````python
# Simple Line Graph
import matplotlib.pyplot as plt

# Data — two simple lists
# x = input values (1 through 5)
# y = output values (doubles of x)
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

# plt.plot() creates a line graph
# scalex=True and scaley=True let Matplotlib auto-fit the axes to your data
# (if False, axes stay fixed regardless of data range)
plt.plot(x, y, scalex=True, scaley=True)

# Add labels and title
plt.title("Simple Line Graph")
plt.xlabel("X values")
plt.ylabel("Y values")

# Display grid — optional but makes it easier to read values
plt.grid(True)

# Display the graph
# NOTE: always call plt.savefig() BEFORE plt.show()
# plt.show() clears the figure, so saving after show gives a blank file
plt.show()
````

````python
# Simple Line Graph
# Use plt.savefig() to save the line graph to a file

import matplotlib.pyplot as plt

# Data
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

# Create line graph
plt.plot(x, y, scalex=True, scaley=True)

# Add labels and title
plt.title("Simple Line Graph")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")

# Save the graph BEFORE showing it
# ⚠️ IMPORTANT: savefig() must come before show()
# plt.show() clears the figure from memory — if you save after, you get a blank file
plt.savefig("line_graph.png")          # saves to current folder
# plt.savefig("G:/full/path/line_graph.png")  # saves to a specific folder path

# Display the graph
plt.show()
````

````python
# Simple Line Graph
# Plot from a pandas DataFrame with markers

import matplotlib.pyplot as plt
import pandas as pd

# Create DataFrame — a table with two columns: month and qty_sold
# This is more realistic than plain lists; real data usually comes in DataFrames
sales = pd.DataFrame({
    "month": [1, 2, 3, 4, 5, 6],
    "qty_sold": [45, 34, 62, 55, 69, 72]
})

# Plot line graph with markers
plt.plot(
    sales["month"],     # x-axis: use the "month" column
    sales["qty_sold"],  # y-axis: use the "qty_sold" column
    marker="o"          # adds a dot (circle) on each data point
                        # try other markers: "s"=square, "^"=triangle, "*"=star
)

# Title and labels
plt.title("Sales at 6 Months")
plt.xlabel("Month")
plt.ylabel("Quantity Sold")

# Save BEFORE show (same rule as above)
plt.savefig("line_graph2.png")

# Show grid (optional but helpful for reading values)
plt.grid(True)

# Display plot
plt.show()
````

````python
# Styling a Pandas line plot using Matplotlib, including:
# marker size (markersize)
# marker color
# line color + style   # fmt shorthand (format string)
# figure size (figsize)

import matplotlib.pyplot as plt
import pandas as pd

# DataFrame
sales = pd.DataFrame({
    "month": [1, 2, 3, 4, 5, 6],
    "qty_sold": [45, 34, 62, 55, 69, 72]
})

# Create figure size (width, height in inches)
# bigger numbers = bigger chart
plt.figure(figsize=(4, 2))

# Plot with styling
plt.plot(
    sales["month"],              # x-axis
    sales["qty_sold"],           # y-axis

    # marker styling
    marker="o",                  # shape: circle (see markers table above)
    markersize=10,               # size of markers (in points)
    markerfacecolor="red",       # fill color of the marker
    markeredgecolor="black",     # border color around the marker

    # line styling
    linestyle="--",              # dashed line (options: "-", "--", "-.", ":")
    linewidth=2,                 # thickness of the line
    color="blue"                 # line color
)

# Labels and title
plt.title("Monthly Sales Trend (Styled)")
plt.xlabel("Month")
plt.ylabel("Quantity Sold")

# Grid
plt.grid(True)

plt.savefig("line_graph3.png")

# Show plot
plt.show()
````

### fmt Parameter (Shortcut Style)

Instead of writing out `marker=`, `color=`, and `linestyle=` separately, you can combine them into one short string called **fmt**:

"ro--"  →  r = red | o = circle marker | -- = dashed line

**Format**: `"[color][marker][linestyle]"`

````python
# fmt Parameter (Shortcut Style)
# Instead of writing marker="o", color="red", linestyle="--" separately,
# you can combine them into one string: "ro--"
# Format: "[color][marker][linestyle]"
# "r" = red, "o" = circle marker, "--" = dashed line

plt.plot(
    sales["month"],
    sales["qty_sold"],
    "ro--",       # fmt = red circles + dashed line (shortcut for color+marker+linestyle)
    markersize=8  # you can still add extra styling on top of fmt
)
````

````python
# Multiple series line graph
# Plotting two lines on the same chart using plt.legend() to tell them apart

import matplotlib.pyplot as plt
import pandas as pd

# Create DataFrame with multiple series (two stores)
sales = pd.DataFrame({
    "month":   [1, 2, 3, 4, 5, 6],
    "store_A": [45, 34, 62, 55, 69, 72],
    "store_B": [30, 40, 38, 46, 60, 65]
})

# Set figure size
plt.figure(figsize=(4, 3))

# Plot multiple lines from DataFrame
# each plt.plot() call adds one line to the chart
# lw = linewidth (shorthand), label = text that appears in the legend
plt.plot(sales["month"], sales["store_A"], marker="o", lw=3, label="Store A")
plt.plot(sales["month"], sales["store_B"], marker="x", lw=3, linestyle="--", label="Store B")

# Titles and labels
plt.title("Monthly Sales Comparison")
plt.xlabel("Month")
plt.ylabel("Quantity Sold")

# Grid + Legend
plt.grid(True)
plt.legend()   # shows the label for each line — needed when you have multiple lines!

plt.savefig("line_graph4.png")

# Show plot
plt.show()
````

### Bar Charts

````python
# Bar Chart from a Pandas DataFrame

import matplotlib.pyplot as plt
import pandas as pd

# Create DataFrame
sales = pd.DataFrame({
    "month":    ["Jan", "Feb", "Mar", "Apr"],
    "qty_sold": [45, 34, 62, 55]
})

# Figure size
plt.figure(figsize=(5, 3))

# Bar chart using plt.bar()
plt.bar(
    sales["month"],       # x-axis (categories)
    sales["qty_sold"],    # y-axis (values)

    color="skyblue",      # bar color
    edgecolor="black",    # border color around each bar
    width=0.6             # bar width (0 to 1; default is 0.8)
)

# Title and labels
plt.title("Monthly Sales (Bar Chart)")
plt.xlabel("Month")
plt.ylabel("Quantity Sold")

# Grid (y-axis only for clarity — avoids visual clutter on x-axis)
plt.grid(axis="y", linestyle="--", alpha=.7)
# alpha controls transparency (0 = invisible, 1 = fully solid)

plt.savefig("bar_graph1.png")

# Show plot
plt.show()
````

### Scatter Plots

## Scatter Charts (`plt.scatter()`)

**Purpose:** Places individual dots on a grid to show the relationship between two numeric variables.
Used to look for trends or clusters — if dots form a diagonal pattern, a relationship exists. If random, no relationship.

**Syntax:**
```python
plt.scatter(x, y, s=size, c=color_array, cmap='colormap_name')
plt.colorbar()
```

| Parameter | Controls | Required? |
|-----------|----------|-----------|
| `x` | horizontal position of each dot | ✅ yes |
| `y` | vertical position of each dot | ✅ yes |
| `s=` | dot size | ❌ no |
| `c=` | dot color | ❌ no |
| `cmap=` | color scheme applied to `c=` values | ❌ no |

**Notes:**
- `c=` and `cmap=` always work as a pair — `cmap=` has nothing to colorize without `c=`
- `plt.colorbar()` adds a color scale legend so you can read what each color represents
- Dot sizes must be **positive numbers** — if using `rand()`, scale up by multiplying since it generates values between 0–1
- You can scale by **any positive number** — the multiplier just controls how big the dots appear:
    - `* 50` → small dots
    - `* 100` → medium dots
    - `* 500` → large dots

**Common colormaps:**
- `'viridis'` → purple → green → yellow
- `'plasma'` → purple → pink → yellow
- `'winter'` → blue → green
- `'coolwarm'` → blue → white → red

````python
# Scatter Plot
# Best used to show the relationship between two numeric variables
# Example: does studying more hours lead to higher exam scores?

import matplotlib.pyplot as plt
import pandas as pd

# Create DataFrame
data = pd.DataFrame({
    "hours_studied": [1, 2, 3, 4, 5, 6, 7],
    "exam_score":    [50, 55, 65, 70, 75, 85, 95]
})

# Figure size
plt.figure(figsize=(4, 2))

# Scatter plot
plt.scatter(
    data["hours_studied"],   # x-axis
    data["exam_score"],      # y-axis

    color="blue",            # point color
    s=100,                   # marker size (NOTE: for scatter use s=, NOT markersize=)
    marker="o",              # marker shape
    alpha=0.7,               # transparency (0=invisible, 1=solid; 0.7 = slightly see-through)
    edgecolors="black"       # border around points
)

# Title and labels
plt.title("Study Hours vs Exam Score")
plt.xlabel("Hours Studied")
plt.ylabel("Exam Score")

# Grid
plt.grid(True)

plt.savefig("scatter1.png")

# Show plot
plt.show()
````

### Histograms

````python
# Histogram
# Shows how frequently values fall within certain ranges (called "bins")
# Different from a bar chart: bar chart compares categories,
# histogram shows distribution of ONE numeric variable

import matplotlib.pyplot as plt
import pandas as pd

# Create DataFrame
data = pd.DataFrame({
    "exam_scores": [45, 55, 62, 70, 75, 80, 85, 90, 95, 60, 68, 73, 77, 88, 92]
})

# Figure size
plt.figure(figsize=(4, 2))

# Histogram
plt.hist(
    data["exam_scores"],   # data column
    bins=5,                # number of intervals (bars) — try changing this!
    color="skyblue",       # bar color
    edgecolor="black",     # border color
    alpha=0.8              # transparency
)

# Title and labels
plt.title("Distribution of Exam Scores")
plt.xlabel("Score Range")
plt.ylabel("Number of Students")

# Grid (y-axis only for readability)
plt.grid(axis="y", linestyle="--", alpha=0.7)

plt.savefig("histogram.png")

# Show plot
plt.show()
````

### Pie Charts

````python
# Pie Chart
# Best for showing percentage share of a whole
# Use when you have ONE data series and want to show how each part contributes

import matplotlib.pyplot as plt
import pandas as pd

# Create DataFrame
data = pd.DataFrame({
    "department": ["IT", "HR", "Finance", "Marketing", "Sales"],
    "employees":  [30, 10, 20, 15, 25]
})

# Figure size (square works best for pie charts)
plt.figure(figsize=(4, 4))

# Pie chart
plt.pie(
    data["employees"],           # values (slice sizes)
    labels=data["department"],   # category names shown on each slice

    autopct="%1.1f%%",           # shows percentage on each slice (e.g. "25.0%")
    startangle=90,               # rotate start position (0=right, 90=top)
    colors=["skyblue", "lightgreen", "orange", "pink", "violet"],
    explode=[0.05, 0, 0, 0, 0],  # highlight first slice by pulling it out slightly
    shadow=True                  # adds a 3D shadow effect
)

# Title
plt.title("Employee Distribution by Department")

plt.savefig("piechart.png")

# Show plot
plt.show()
````
