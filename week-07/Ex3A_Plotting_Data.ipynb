# Exercise 3.A Getting Familiar with Matplotlib

```python
import numpy as np               # Importing for array operations and data manipulation
import matplotlib.pyplot as plt  # Importing for creating visualizations and plots

# A command that displays plots directly in the notebook
%matplotlib inline              
```

## Creating Basic Plots

```python
# Creating two variables for data points
x = [1, 2, 3, 4, 5] 
y = [2, 3, 5, 7, 11] 
```

```python
# Creating and Displaying the line graph

plt.plot(x, y) # .plot method creates a line graph
plt.show()     # Displays the graph
```

```python
# Adding markers to data points in to original line graph

plt.plot(
    x, y,
    marker="o" # Adds a circle on each data point
)

plt.show() # Displays new line graph with markers
```

```python
# Customizing labels in previous line graph (title and axis labels)

plt.plot(
    x, y,
    marker="o" # Adds a circle on each data point
)

# Title and labels
plt.title("Example Graph")
plt.xlabel("X axis")
plt.ylabel("Y axis")

plt.show() # Displays new line graph with markers and labels
```

## Styling Plots

```python
# Customizing Line Style

plt.plot(
    x, y, 
    marker="o",
    linestyle="--",   # Styles the line as a dashed lin
    color="red"       # Assigns the line a red color
)

plt.title("Example Graph")
plt.xlabel("X axis")
plt.ylabel("Y axis")

# Display new graph
plt.show()
```

```python
# Using shorcut to customize line style (fmt parameter)
# Format: "[color][marker][linestyle]"
# "r" = red color, "s" = square marker, "--" = dashed line
# fmt string ('rs--') quickly sets line color, marker shape, and line style
# ms and mfc are keyword arguments that fine-tune details fmt can't handle separately 

plt.plot(x, y, 'rs--', ms = 10, mfc = 'g') 
# ms = marker size
# mfc = markerface color

plt.title("Example Graph")
plt.xlabel("X axis")
plt.ylabel("Y axis")

plt.show()
```

```python
# Example 1: Experimenting with colors

plt.plot(x, y, 'D:', ms = 8, mfc = 'm', color='pink')
# some colors like pink have no shorcut, so the full color name is required
plt.title("Example Graph")
plt.xlabel("X axis")
plt.ylabel("Y axis")

plt.show()
```

```python
# Example 2: Experimenting with colors
plt.plot(x, y, 'b^-.', ms = 10, mfc = 'pink') 
# ms = marker size
# mfc = markerface color

plt.title("Example Graph")
plt.xlabel("X axis")
plt.ylabel("Y axis")

plt.show()
```
