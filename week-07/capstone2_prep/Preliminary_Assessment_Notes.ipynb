#### **Section 1.3** Merging 

1. What does `pd.merge()` do?
- This function merges matching or similar data from more than one dataset based on a shared key, which is often some sort of ID (StoreID, SaleID, ProductID, etc) to get information from another table. Similar to SQL's join, it includes inner joins, outer joins, left joins and right joins.

2. What is the difference between `on=` and `left_on= / right_on=`?
- on= and left_on, right_on are about which columns to join on, not which table. on= is used when both tables share the same column name. left_on/right_on is used when column names are different, but share the same kind of data you want, two columns with customer names can both share the same kind of information (customer names) but be named under a different column. 


3. What is the difference between an inner join and left join?
- An inner join keeps matching rows from both tables, while left join keeps all rows from the left table, or first table only. It matches what it can from the right table, but if there's no match, it fills with NaN values.

#### **Section 2.3** Sorting and Ranking

- This method sorts values in a column by a specified order. The two most parameters you'd pass into it are the name of the column, and the order, either ascending=True (smallest to largest, default) or ascending=False (descending, largest to smallest). There is no descending=True option, since ascending is the default.
      
2. What is the difference between `.head()` and `.nlargest(10)`? When would you use each?
- .head(10) returns the top 10 rows of the DataFrame, while n.largest(10) returns the 10 largest numerical values. Use .head() after sorting, and .nlargest() does the sorting step and returning largest numerical values step at the same time

3. You want to find the top 5 performing stores by total revenue in Texas. You already have a grouped DataFrame called texas_revenue with columns Store ID and Sale Amount. What is the exact sequence of steps to get your answer?
- groupby Store ID and sum Sale Amount, sort by sale amount descending, take top 5
- texas_revenue = store_sales.groupby('Store ID')['Sale Amount'].sum().reset_index()

#### **Section 2.4** Pivot Tables

1. In your own words, what does `pd.pivot_table()` do differently than groupby()?
- The main thing pd.pivot_table does differently than groupby is the way it returns results. Both functions can group columns based on similar values and perform aggregate operations on each row, but groupby returns results in a long/list format, while pivot_table returns results  as a table. When a combination doesn't exist in the data, those missing values becom NaN in the pivot table, so .fillna(0) is needed to replace NaN values so the charts display and work properly.
    
2. Look at this syntax and explain what each parameter does:

```python
pd.pivot_table(store_sales,          # DataFrame to use
               index='Category',     # groups the ROWS
               columns='State',      # groups the COLUMNS
               values='Sale Amount', # what is being calculated
               aggfunc='sum')        # what operation to perform
```
               
3. Why would you need .fillna(0) after creating a pivot table? What situation causes NaN values to appear?
- I need .fillna(0) after creating a pivot table to account for possible missing data that can cause NaN values to appear in the table.
