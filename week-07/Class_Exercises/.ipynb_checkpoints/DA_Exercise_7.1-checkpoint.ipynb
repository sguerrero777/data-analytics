This notebook uses NumPy fundamentals to load, inspect, and analyze data. It covers array creation, descriptive statistics, element-wise arithmetic, Boolean masking, and 2-D reshaping for five regional offices at SalesMetrics Co.

```python
import numpy as np

# Creating the 1D array revenue to store weekly revenue values and 
# printing its type and atrributes
revenue = np.array([52400.0, 38750.0, 61200.0, 47900.0, 55650.0])

print("Array                       :", revenue)
print("Data Type of Revenue Array  :", revenue.dtype) 
# float64 because numbers have decimals, and each number uses 64 bits of memory 
# for high precision. If values were whole numbers with no decimal, it would be int64
print("Attributes of Revenue Array :", revenue.shape)
# Only one dimension with 5 values, no rows or columns
```

```python
# Descriptive Statistics with NumPy's Built-in Functions

print("========== Revenue Statistics ==========")
print(f"Total      : ${np.sum(revenue):,.2f}")
print(f"Mean       : ${np.mean(revenue):,.2f}")
print(f"Median     : ${np.median(revenue):,.2f}")
print(f"Std Dev    : ${np.std(revenue):,.2f}")
print(f"Minimun    : ${np.min(revenue):,.2f}")
print(f"Maximum    : ${np.max(revenue):,.2f}")
```

```python
# Element-wise Arithmetic Functions to Calculate Bonuses

revenue_bonus = np.multiply(revenue, 1.05) # np.multiply() requires two arguments
# multiply by 1.05 to give full original plus five percent bonus
bonus_amount = np.subtract(revenue_bonus, revenue)

print("Original Revenue       :", revenue)
print("Revenue After 5% Bonus :", revenue_bonus)
print("Bonus Amount           :", bonus_amount)

# NumPy applies the operation to every element, no loop needed!
```

```python
# Boolean Masking to Filter High-Performing Regions

above_avg = revenue > np.mean(revenue) 
# Applys a condition to check every element and return an array where the condition is true
values = revenue[above_avg]
# Filters and stores only the actual revenue values where the condition was True
regions_above_avg = np.sum(above_avg)
# Stores a sum of the elements where the condition was true

print("Mask True/False        :", above_avg)
print("Above Average          :", values) 
print(f"Regions above average  : {regions_above_avg}")
```

```python
# Reshape and 2D Analysis
# Reshaping allows us to view the data as a matrix and compute axis-based stats.

print("Column Array:", revenue.reshape(5, 1))
# Reshapes data to a table format with 5 rows and 1 column for stacking

# vstack() stacks vertically along row axis
# Creating a 2D array by stacking revenue and revenue with 5% bonus
# to simulate two weeks of data — Week 1 original, Week 2 after bonus
q_data = np.vstack([revenue, revenue * 1.05])
print("\n2D Array (2 weeks x 5 regions):\n", q_data)
print("\nRevenue 2D:", q_data)

print("\nWeekly Totals    :", np.sum(q_data, axis=1))
print("Region average   :", np.mean(q_data, axis=0))
# axis=1 indicates across columns (left to right), so it adds up values for all five regions for each week
# specifically, it takes all the column values in one row and adds them, doesnt' mix with other rows
# axis=0  moves across rows (top to bottom), so it gives 2-week average per region
# specifically, it goes down each column to find the means, mixing the weeks to find each region's average across two different performance weeks
```

## Region Analysis Summary

- The five regions generated a **combined total** of `$255,900` in Week 1 and `$268,695` in Week 2 after sales performance bonuses.
- **Region 3** is the top perfomer at `$61,200` in sales, with Region 2 the lowest performer at `$38,750`
