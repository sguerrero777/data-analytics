# Exercise 2.C Working with Multiple DataFrames

```python
import numpy as np
import pandas as pd
```

## Basic Aggregations

```python
# Creating the DataFrame from dataset
data = {
    'Product': ['A', 'B', 'C', 'A', 'B', 'C', 'A', 'B', 'C'],
    'Region': ['North', 'South', 'East', 'West', 'North', 'South', 'East', 'West', 'North'],
    'Units_Sold': [10, 15, 20, 5, 10, 15, 10, 5, 15],
    'Price': [100, 200, 150, 100, 200, 150, 100, 200, 150]
}

df_sales = pd.DataFrame(data)
df_sales
```

```python
# Adding new column called Revenue to DataFrame
df_sales["Revenue"] = df_sales['Units_Sold'] * df_sales['Price']

print("\n******** Updated Sales Table ********\n")
print(df_sales)

# Total Units_Sold and total Revenue across all products and regions
# Double brackets are used because the outer [] is the DataFrame indexing operator
# and the inner [] is a Python list of column names
# Single brackets select one column; double brackets select multiple columns as a list\

print("\n Total Units Sold | Total Revenue")
print(df_sales[['Units_Sold', 'Revenue']].sum())

# Average Units_Sold and average Revenue
print("\n Average Units Sold | Average Total Revenue")
print(df_sales[['Units_Sold', 'Revenue']].mean())
```

## Grouping Data

```python
# Using groupby to group df_sales by Product and calculate the total Units_Sold and total
# revenue for each product
# groupby splits DataFrame into groups based on a column, then applying an aggregation function
# df.groupby('column_name')['column_to_aggregate'].function()
# groupby() groups all rows that share the same value in a column together
# Unlike .sum() alone which gives one total for everything, groupby keeps groups separate
# Instead of 9 individual rows, we get 3 rows — one per product (A, B, C)
# .sum() then adds up the Units_Sold and Revenue for each product individually
df_sales.groupby('Product')[['Units_Sold', 'Revenue']].sum()
```

```python
# Using groupby() to group df_sales by Region and calculate the average Units_Sold and
# and average Revenue for each region.
# Splitting Database by Region column, and for each region calculating the average Units_Sold per
# Region and average revenue per region
df_sales.groupby('Region')[['Units_Sold', 'Revenue']].mean()
#.round(2) can be used to round
```

```python
# Using groupby to to group df_sales by both Product and Region and calculate the sum of 
# Units_Sold and revenue for each combo
# Splitting Database by Region and Product column, so I'll get the sum for each product sold 
# in each region and the revenue for each region based on that product
df_sales.groupby(['Product', 'Region'])[['Units_Sold', 'Revenue']].sum()\
# When grouping by multiple columns, they must be in a list
```

## Sorting and Ranking Data

```python
# Using sort_values() method to sort Units_Sold in ascending order
# df.sort_values("col")
print("----- Lowest to Highest Units Sold -----\n")
sorted_df_sales = df_sales.sort_values('Units_Sold', ascending=True)
print(sorted_df_sales)
# ascending is the default so you can technically leave it empty)
```

```python
# Using sort_values() method to sort Revenue in descending order
print("----- Individual Sales Entries by Highest Revenue  -----\n")
sorted_df_revenue = df_sales.sort_values('Revenue', ascending=False)
print(sorted_df_revenue)
# descending can never be a paramter because the default is always ascending, so it must 
# either be True or False. Use = False for descending
```

```python
# Using .rank() method to rank Revenue values in descending order in a new column
# Highested revenue = rank 1
# Syntax: df["col"].rank()
df_sales["Revenue_Rank"] = df_sales['Revenue'].rank(ascending=False)
df_sales

# Note: rows 1 and 2 both have Revenue 3000 (a tie), so pandas averages ranks 1 and 2
# giving both a rank of 1.5 instead of rank 1 — this is pandas' default tie-breaking behavior
```

## Merging DataFrames

```python
# Creating new DataFrame for Products

product_data = {
    'Product': ['A', 'B', 'C'],
    'Category': ['Electronics', 'Furniture', 'Appliances']
}
df_products = pd.DataFrame(product_data)
df_products
```

```python
# Merge df_sales with df_products on the Product column, resulting in a new DataFrame named df_merged
# # The purpose of this merge is to add the Category column from df_products to df_sales
# Both DataFrames share the Product column which is used as the key to match rows
df_merged = pd.merge(df_sales, df_products, on="Product", how="inner")
df_merged
# Using inner merge since all products (A, B, C) exist in both DataFrames
# There are no unmatched rows so all merge types produce the same result here
# Inner is the standard choice when all keys match between both DataFrames
# Outer merge would only be needed if there were unmatched rows to preserve
```

```python
# Using groupby on df_merged to calculate the total revenue by category
# I'm grouping by category, and calculating the sum revenue on each category
df_merged.groupby('Category')[['Revenue']].sum()
```
