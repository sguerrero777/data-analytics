# Exercise 2.A Fundamentals of Pandas 

```python
import numpy as np
import pandas as pd
```

## Series

```python
# Creating Variables to hold plain Python Objects
aa = [1, 2, 3, 4]
ab = (5, 6, 7, 8)
ac = {9, 10, 11, 12}
ad = {
    "first": 13,
    "second": 14,
    "third": 15,
    "fourth": 16
}

print(f"List Object: {aa}\n")
print(f"Tuple Object: {ab}\n")
print(f"Set Object: {ac}\n")
print(f"Dictionary Object: {ad}\n")
```

```python
# Creating Series from  Python Objects
# A Pandas Series is a one-dimensional labeled array
# pd.Series converts Python objects into Pandas Series
# Series are index-based
ser1 = pd.Series(aa)
print(f"List to Series: \n{ser1}\n")

ser2 = pd.Series(ab)
print(f"Tuple to Series: \n{ser2}\n")

# ser3 = pd.Series(ac)
# since Pandas can't convert a set to a series, it must be converted into a list or tuple
# I created a new variable called modifies_ser3 to convert the result of the list 
# conversion to a series
modified_ser3 = pd.Series(list(ac))
print(f"Set to Series: \n{modified_ser3}\n")

ser4 = pd.Series(ad)
print(f"Dictionary to Series: \n{ser4}\n")
```

```python
ser5 = pd.Series(17, 18, 19, 20)

# The type error that arises is " Index(...) must be called with a collection of some kind"
# This means that you cannot covert a series without first assigning a collection of elements, 
# wether it be a list, dictionary, or tuple. The reason for this is that these collections
# have index positions; since Pandas series are index-based and literal values don't have
# indexes, Pandas can't convert them to a series.
```

## DataFrames from Series

```python
# Creating a DataFrame from Series
df1 = pd.DataFrame(ser1)
df1 # No print function is necessary, shows a cute table
```

```python
# Inserting New Columns to a DataFrame using .insert() method
# Format inserting from series and DataFrames is the same
# Format is index position of where you want to insert new column, column name, and column data
df1.insert(1, "ser2", ser2)
df1
```

```python
# .insert() only inserts one column at a time
# Inserting data from modified series 3
df1.insert(2, "ser3", modified_ser3)
df1
```

```python
df1.insert(3, "ser4", ser4)
df1
# ser4 is a dictionary-based Series, so its index is made up of string keys
# ('first', 'second', 'third', 'fourth') instead of numeric positions (0, 1, 2, 3).
# Since df1 uses a numeric index, pandas cannot align ser4's string index with it,
# resulting in NaN values across the entire column.
```

```python
# Using the loc attribute to add a new row
# In a DataFrame, you can add new rows by assigning to a loc index that doesn't exist yet
# by setting the data, it will just extend the dataframe. You can use either lists or tuples
df1.loc[4] = [5, 9, 12, 0]
df1
```

```python
# Using .describe() to understand distribution and stats of DataFrame
df2.describe()
```

#### Useful Insights from `.describe()`
Since the sakila dataset contains text-based (object) columns, `.describe()` returns 
categorical statistics rather than numeric ones. It's showing count, unique values, the most 
frequent value (top), and its frequency (freq).

A useful insight is the top category and its frequency, which tells us which genre appears most often in the dataset. Another useful insight is the count and unique values, which show how many records exist and how much variety there is across titles and actors.

A less useful insight is the top actor_name, because it only shows which actor appears most frequently in the dataset — not how many unique movies they appear in or whether their films perform well commercially.

```python
# Using the read_csv() function to read contents of northwind database and convert to a DataFrame
df3 = pd.read_csv("northwind_example.csv")
df3
```

```python
# Using .describe() to understand distribution and stats of DataFrame
df3.describe()

# 25%, 50%, and 75% are percentiles (quartiles) that show the distribution of the data
# 25% (first quartile) — 25% of the data falls below this value
# 50% (second quartile/median) — half the data falls below this value
# 75% (third quartile) — 75% of the data falls below this value
# Example: for Quantity, 50% of orders had a quantity of 20 or less, and 75% had 32 or less
```

#### Useful Insights from `.describe()`
Since the northwind dataset contains numerical columns, .describe() returns numerical statistics rather than categorical ones like the sakila database. It's showing the count, mean, standard deviation, minimum, maximum, and percentages across OrderID, UnitPrice, and Quantity.

A useful insight is the average quantity sold, minimum, and maximum quantity sold because it can provide indicators for top and lowest performing products.

A less useful insight is the OrderID statistics (mean, std, min, max) because OrderID is just an identifier, not a meaningful numeric value. Calculating statistics on OrderID doesn't reveal anything useful about the business data.

## Coding a DataFrame

```python
# Creating a new variable and assiging it a dictionary
test_data = {
    'ID': [1, 2, 3, 4, 5, 6, 7, 8],
    'Name': ['Alicia', 'Bob', 'Charlie', 'Dominic', 'Eve', 'Frank', 'Grace', 'Heidi'],
    'Age': [25, np.nan, 22, 19, 31, 35, np.nan, 18],
    'City': ['NY', 'LA', 'Chicago', 'Houston', 'Phoenix', 'Boston', 'Austin', 'San Diego'],
    'Score': [85, 92, -1, 65, 78, np.nan, 55, 90]
}
```

```python
# Creating a DataFrame from a dictionary
df4 = pd.DataFrame(test_data)
df4
```

```python
# Exporting the DataFrame to a csv file
# You can export to csv, excel, or json file
# index=false is a boolean parameter that excludes the pandas index, making sure your 
# exported file doesn't have it as an extra unwanted column
df4.to_csv('test_data.csv', index=False)
```

The export was successful, I now have a test_data.csv file in my week-07 folder. It looks clean and the NaN values are just empty cells, no errors!
