# Exercise 2.B Cleaning up Data 

```python
import numpy as np
import pandas as pd
```

```python
# Creating a DataFrame using csv file
df = pd.read_csv("test_data.csv")
df
```

```python
# Showing column statistics using method
df.describe()
```

## Filtering a DataFrame

```python
# Filtering and displaying rows where Age is greater than 18
filter_age = df[df['Age'] > 18]
print(filter_age)

# the inner part df['Age'] > 18 is a filter/mask that stores the TRUE/FALSE series where the condition is met
# the outer brackets take that series and return the actaul rows where the value is true
```

```python
# Filtering and Displaying rows where Score is greater than 50 and less than 100
# When you have 2 conditions, you need to use & and  wrap each condition in its own ()
# Pandas uses & for AND and | for OR
# Both have to be true regardless of which side they're on
filter_score = df[(df['Score'] > 50) & (df['Score'] < 100)]
print(filter_score)
```

## Modifying a DataFrame

```python
# Creating a new variable that stores a copy of the DataFrame but cleaned. Original is untouched
# Using the drop function to remove rows where Score is invalid(less than 0)
# drop is for intentional data
# dropna is for all missing values
df_cleaned = df.drop(df[df['Score'] < 0].index)
# df.drop() combines boolean indexing and a condition to remove specific rows
# The inner mask df['Score'] < 0 creates a True/False Series identifying where the condition is met
# The outer wrap df[df['Score'] < 0] grabs the actual rows where the condition is True
# .index extracts the index labels of those rows
# .drop() then knows and removes the rows with those index labels from the DataFrame

df_cleaned
```

```python
# Dropping a city column
df_cleaned = df_cleaned.drop("City", axis = 1)
# axis=1 targets columns because the operation moves horizontally across the DataFrame to find them
# axis=0 targets rows because the operation moves vertically down the DataFrame to find them
# drop() takes the column/row name and axis direction as parameters
# Think of a grid, 0 is vertical axis and 1 is horizontal axis
# drop can remove columns with column name and the the axis postion as parameters
df_cleaned
```

```python
# Reaplacing missing Age values with the mean of Age in DataFrame
df_cleaned['Age'] = df_cleaned['Age'].fillna(df_cleaned['Age'].mean())
df_cleaned
# .fillna replaces missing values
# I want the mean of the cleaned DataFrame because it has no invalid rows
# ['Age'] specifies and targets the Age column
```

```python
# Replacing score values of 0 with the mean Score using replaced()
# .replace() replaces specific values
# It needs two arguments, what to replace and what to replace it with
df_cleaned['Score'] = df_cleaned['Score'].replace(0, df_cleaned['Score'].mean())
df_cleaned
```

```python
# Removing any NaN values using .dropna()
df_cleaned = df_cleaned.dropna()
df_cleaned

# We need to remove it to avoid unreliable resujlts when doing analysis or calculations since there's no score
```

```python
# Setting the ID column as the index labels
# set_index() moves the ID column out of the data and into the index, removing it automatically
df_cleaned = df_cleaned.set_index('ID')
df_cleaned

# Here we replace the default numeric index with the actual ID values
# Setting ID as the index makes lookups more easier. Instead of searching by row number,
# you can search by the actual record ID which maps directly to a real entity.
# It also makes merging DataFrames easier when joining on a shared ID.
```
